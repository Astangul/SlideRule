```python
import numpy as np
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt

def fit_data_dist_thick_new_equation(xdata, thickness_data, ydata, plot=True):
    """
    Ajuste le modèle :
        y = A * x^(-(a*thickness + b)) * exp(-mu_air*x) * exp(-mu_wall*thickness)
    
    Renvoie :
    - A_opt, a_opt, b_opt, mu_air_opt, mu_wall_opt : paramètres optimaux
    - R_squared : coefficient de détermination (calculé sur ln(y))
    - uncertainties : incertitudes sur [A, a, b, mu_air, mu_wall].
    """

    # --------------------------
    # 1) Filtrage des données
    # --------------------------
    xdata = np.array(xdata, dtype=float)
    thickness_data = np.array(thickness_data, dtype=float)
    ydata = np.array(ydata, dtype=float)

    valid = (~np.isnan(ydata)) & (ydata > 0) & (xdata > 0)
    xdata = xdata[valid]
    thickness_data = thickness_data[valid]
    ydata = ydata[valid]

    if len(ydata) == 0:
        raise ValueError("Aucune donnée valable pour l'ajustement.")

    # --------------------------
    # 2) Passage en log(y)
    # --------------------------
    ln_ydata = np.log(ydata)

    # --------------------------
    # 3) Construction de la matrice X
    #
    # On a :
    # ln(y) = ln(A)
    #         - b ln(x)
    #         - a [thickness * ln(x)]
    #         - mu_air * x
    #         - mu_wall * thickness
    #
    # On l'écrit : ln(y) = Beta0 + Beta1 ln(x) + Beta2 [th ln(x)] + Beta3 x + Beta4 th
    # en remarquant que :
    #   Beta0 = ln(A)
    #   Beta1 = -b
    #   Beta2 = -a
    #   Beta3 = -mu_air
    #   Beta4 = -mu_wall
    # --------------------------
    ln_x = np.log(xdata)
    # Colonne 1 : 1
    # Colonne 2 : ln(x)
    # Colonne 3 : thickness * ln(x)
    # Colonne 4 : x
    # Colonne 5 : thickness
    X = np.column_stack([np.ones_like(xdata),
                         ln_x,
                         thickness_data * ln_x,
                         xdata,
                         thickness_data])

    # Résolution par moindres carrés linéaire
    Beta, residuals, rank, s = np.linalg.lstsq(X, ln_ydata, rcond=None)
    # Interprétation
    lnA_init = Beta[0]
    b_init = -Beta[1]
    a_init = -Beta[2]
    mu_air_init = -Beta[3]
    mu_wall_init = -Beta[4]

    initial_guess = [lnA_init, a_init, b_init, mu_air_init, mu_wall_init]

    # --------------------------
    # 4) Définition de la fonction "log_model" pour curve_fit
    #    On suit l'ordre [lnA, a, b, mu_air, mu_wall]
    # --------------------------
    def log_model(vars, lnA, a, b, mu_air, mu_wall):
        x, th = vars
        # y = A * x^{-(a*th + b)} * exp(-mu_air*x) * exp(-mu_wall*th)
        # ln(y) = ln(A) - (a*th + b)*ln(x) - mu_air*x - mu_wall*th
        return (lnA
                - (a*th + b)*np.log(x)
                - mu_air*x
                - mu_wall*th)

    # --------------------------
    # 5) Ajustement final via curve_fit
    # --------------------------
    popt, pcov = curve_fit(
        log_model,
        (xdata, thickness_data),
        ln_ydata,
        p0=initial_guess,
        maxfev=10000
    )

    lnA_opt, a_opt, b_opt, mu_air_opt, mu_wall_opt = popt
    A_opt = np.exp(lnA_opt)

    # Incertitudes
    perr = np.sqrt(np.diag(pcov))  # [lnA, a, b, mu_air, mu_wall]
    A_unc = A_opt * perr[0]
    a_unc = perr[1]
    b_unc = perr[2]
    mu_air_unc = perr[3]
    mu_wall_unc = perr[4]
    uncertainties = [A_unc, a_unc, b_unc, mu_air_unc, mu_wall_unc]

    # --------------------------
    # 6) Calcul de R^2 (en log)
    # --------------------------
    ln_yfit = log_model((xdata, thickness_data), *popt)
    ss_res = np.sum((ln_ydata - ln_yfit)**2)
    ss_tot = np.sum((ln_ydata - np.mean(ln_ydata))**2)
    R_squared = 1 - ss_res / ss_tot

    # --------------------------
    # 7) Tracé optionnel
    # --------------------------
    if plot:
        plt.figure(figsize=(8, 6))
        plt.title("Fit : y = A * x^(-(a*thickness + b)) * ...")

        # Palette de couleurs
        unique_th = np.unique(thickness_data)
        import matplotlib.cm as cm
        colors = cm.viridis(np.linspace(0, 1, len(unique_th)))

        for th_val, col in zip(unique_th, colors):
            mask = (thickness_data == th_val)
            x_sub = xdata[mask]
            y_sub = ydata[mask]

            # Points expérimentaux
            plt.loglog(x_sub, y_sub, 'o', color=col, label=f"ép={th_val} mm (data)")

            # Courbe ajustée
            x_fit = np.logspace(np.log10(x_sub.min()), np.log10(x_sub.max()), 100)
            # y_fit = A * x_fit^{-(a*th_val + b)} * exp(-mu_air*x_fit) * exp(-mu_wall*th_val)
            y_fit = (A_opt
                     * x_fit**(-(a_opt*th_val + b_opt))
                     * np.exp(-mu_air_opt*x_fit)
                     * np.exp(-mu_wall_opt*th_val))
            plt.loglog(x_fit, y_fit, '-', color=col, label=f"ép={th_val} mm (fit)")

        plt.xlabel("Distance (m)")
        plt.ylabel("Dose (Gy)")
        plt.legend()
        plt.grid(True, which="both", ls="--")

        formula_text = (
            f"y = {A_opt:.3g} · x^(-[{a_opt:.3g}·th + {b_opt:.3g}]) "
            f"· exp(-{mu_air_opt:.3g}·x) · exp(-{mu_wall_opt:.3g}·th)\n"
            f"R²(log) = {R_squared:.4f}"
        )
        plt.text(0.05, 0.05, formula_text, transform=plt.gca().transAxes,
                 bbox=dict(facecolor='white', alpha=0.8), fontsize=10)
        plt.show()

    # --------------------------
    # 8) Retour
    # --------------------------
    return A_opt, a_opt, b_opt, mu_air_opt, mu_wall_opt, R_squared, uncertainties
```

```python
import numpy as np
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt

def fit_dose_dist_lead_buildup_in_screen(r_data, T_data, D_data, plot=True):
    """
    Ajuste le modèle :
        D(r,T) = (A / r^2) * exp(-mu_air * r) * [1 + c*T] * exp(-mu_pb * T)

    Paramètres
    ----------
    r_data : array-like
        Distances (m)
    T_data : array-like
        Épaisseurs d'écran (plomb), en m ou cm (selon mu_pb).
    D_data : array-like
        Doses mesurées (ou intensités).
    plot : bool
        Si True, affiche un graphe log-log comparant les données et le fit.

    Retour
    ------
    A_opt, mu_air_opt, mu_pb_opt, c_opt : float
        Paramètres optimaux.
    R2_log : float
        Coefficient de détermination (calculé sur ln(D)).
    uncertainties : list
        [delta_A, delta_mu_air, delta_mu_pb, delta_c]
    """

    # 1) Filtrage des données invalides
    r_data = np.array(r_data, dtype=float)
    T_data = np.array(T_data, dtype=float)
    D_data = np.array(D_data, dtype=float)

    valid = (~np.isnan(D_data)) & (D_data > 0) & (r_data > 0) & (T_data >= 0)
    r_data = r_data[valid]
    T_data = T_data[valid]
    D_data = D_data[valid]

    if len(D_data) == 0:
        raise ValueError("Aucune donnée valable pour l'ajustement.")

    # 2) Logarithme des doses
    ln_D_obs = np.log(D_data)

    # 3) Estimation initiale (en ignorant le build-up) => c=0
    #
    #    On se ramène au modèle "classique" sans build-up :
    #    D ~ (A/r^2) * exp(-mu_air*r) * exp(-mu_pb*T)
    #
    #    ln(D) + 2 ln(r) = ln(A) - mu_air*r - mu_pb*T
    #
    ln_r = np.log(r_data)
    Y = ln_D_obs + 2*ln_r  # côté gauche
    X = np.column_stack([np.ones_like(r_data), r_data, T_data])
    Beta, _, _, _ = np.linalg.lstsq(X, Y, rcond=None)

    lnA_init = Beta[0]
    mu_air_init = -Beta[1]
    mu_pb_init = -Beta[2]

    A_init = np.exp(lnA_init)
    c_init = 0.0  # hypothèse simple : pas de build-up

    initial_guess = [A_init, mu_air_init, mu_pb_init, c_init]

    # 4) Définition de la fonction ln(D_mod) pour curve_fit
    def log_model(vars, A, mu_air, mu_pb, c):
        r, T = vars
        # D_mod = (A / r^2) * exp(-mu_air*r) * [1 + c*T] * exp(-mu_pb*T)
        # ln(D_mod) = ln(A) - 2 ln(r) - mu_air*r + ln(1 + c*T) - mu_pb*T
        return (np.log(A)
                - 2*np.log(r)
                - mu_air*r
                + np.log(1 + c*T)
                - mu_pb*T)

    # 5) Ajustement via curve_fit
    popt, pcov = curve_fit(
        log_model,
        (r_data, T_data),   # x
        ln_D_obs,           # y
        p0=initial_guess,
        maxfev=10000
    )

    A_opt, mu_air_opt, mu_pb_opt, c_opt = popt
    perr = np.sqrt(np.diag(pcov))  # incertitudes
    dA, d_mu_air, d_mu_pb, d_c = perr
    uncertainties = [dA, d_mu_air, d_mu_pb, d_c]

    # 6) Calcul de R² (log-space)
    ln_D_fit = log_model((r_data, T_data), *popt)
    ss_res = np.sum((ln_D_obs - ln_D_fit)**2)
    ss_tot = np.sum((ln_D_obs - np.mean(ln_D_obs))**2)
    R2_log = 1 - ss_res / ss_tot

    # 7) Tracé optionnel
    if plot:
        plt.figure(figsize=(8, 6))
        plt.title("Fit avec build-up dans l'écran")

        unique_T = np.unique(T_data)
        import matplotlib.cm as cm
        colors = cm.viridis(np.linspace(0, 1, len(unique_T)))

        for T_val, col in zip(unique_T, colors):
            mask = (T_data == T_val)
            r_sub = r_data[mask]
            D_sub = D_data[mask]

            # Données expérimentales
            plt.loglog(r_sub, D_sub, 'o', color=col, label=f"T={T_val} (data)")

            # Courbe ajustée (r variable)
            r_min, r_max = r_sub.min(), r_sub.max()
            r_fit = np.logspace(np.log10(r_min), np.log10(r_max), 200)
            D_fit = (A_opt / r_fit**2
                     * np.exp(-mu_air_opt*r_fit)
                     * (1 + c_opt*T_val)
                     * np.exp(-mu_pb_opt*T_val))

            plt.loglog(r_fit, D_fit, '-', color=col, label=f"T={T_val} (fit)")

        plt.xlabel("Distance r")
        plt.ylabel("Dose D")
        plt.legend()
        plt.grid(True, which="both", ls="--")

        formula_text = (
            f"D(r,T) = {A_opt:.3g}/r^2 · exp(-{mu_air_opt:.3g}·r) · [1 + {c_opt:.3g}·T] · exp(-{mu_pb_opt:.3g}·T)\n"
            f"R² (log) = {R2_log:.4f}"
        )
        plt.text(0.05, 0.05, formula_text,
                 transform=plt.gca().transAxes,
                 bbox=dict(facecolor='white', alpha=0.7),
                 fontsize=10)
        plt.show()

    return A_opt, mu_air_opt, mu_pb_opt, c_opt, R2_log, uncertainties
```

```python
import numpy as np
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt

def fit_dose_dist_lead_exponent_build_up(r_data, T_data, D_data, plot=True):
    """
    Ajuste le modèle :
        D(r,T) = A * r^(-n) * (1 + c*T) * exp(-mu_air*r) * exp(-mu_pb*T)

    Paramètres
    ----------
    r_data : array-like
        Distances (m)
    T_data : array-like
        Épaisseurs d'écran (plomb), en m ou cm (selon mu_pb).
    D_data : array-like
        Doses mesurées (ou intensités).
    plot : bool
        Si True, affiche un graphe log-log comparant les données et le fit.

    Retour
    ------
    A_opt, n_opt, c_opt, mu_air_opt, mu_pb_opt : float
        Paramètres optimaux.
    R2_log : float
        Coefficient de détermination (calculé sur ln(D)).
    uncertainties : list
        [delta_A, delta_n, delta_c, delta_mu_air, delta_mu_pb]
    """

    # 1) Filtrer les données invalides
    r_data = np.array(r_data, dtype=float)
    T_data = np.array(T_data, dtype=float)
    D_data = np.array(D_data, dtype=float)

    valid = (~np.isnan(D_data)) & (D_data > 0) & (r_data > 0) & (T_data >= 0)
    r_data = r_data[valid]
    T_data = T_data[valid]
    D_data = D_data[valid]

    if len(D_data) == 0:
        raise ValueError("Aucune donnée valable pour l'ajustement.")

    ln_D_obs = np.log(D_data)

    # 2) Estimation initiale (optionnelle) en supposant n=2, c=0,
    #    on récupère A, mu_air, mu_pb en ignorant c.
    #
    #    D ~ A / r^2 * exp(-mu_air*r) * exp(-mu_pb*T)
    # => ln(D) + 2 ln(r) = ln(A) - mu_air*r - mu_pb*T
    #
    ln_r = np.log(r_data)
    Y = ln_D_obs + 2*ln_r  # côté gauche
    X = np.column_stack([np.ones_like(r_data), r_data, T_data])  # [1, r, T]
    Beta, _, _, _ = np.linalg.lstsq(X, Y, rcond=None)
    lnA_init = Beta[0]
    mu_air_init = -Beta[1]
    mu_pb_init = -Beta[2]
    A_init = np.exp(lnA_init)

    # On part avec n=2, c=0 comme guess
    n_init = 2.0
    c_init = 0.0

    initial_guess = [A_init, n_init, c_init, mu_air_init, mu_pb_init]

    # 3) Définition du modèle en log pour curve_fit
    def log_model(vars, A, n, c, mu_air, mu_pb):
        r, T = vars
        # D = A * r^(-n) * (1 + c*T) * exp(-mu_air*r) * exp(-mu_pb*T)
        # ln(D) = ln(A) - n ln(r) + ln(1 + c*T) - mu_air*r - mu_pb*T
        return (np.log(A)
                - n*np.log(r)
                + np.log(1 + c*T)
                - mu_air*r
                - mu_pb*T)

    # 4) Ajustement via curve_fit
    popt, pcov = curve_fit(
        log_model,
        (r_data, T_data),
        ln_D_obs,
        p0=initial_guess,
        maxfev=10000
    )

    A_opt, n_opt, c_opt, mu_air_opt, mu_pb_opt = popt

    # 5) Incertitudes
    perr = np.sqrt(np.diag(pcov))  # [dA, dn, dc, d_mu_air, d_mu_pb]
    uncertainties = list(perr)

    # 6) Calcul de R^2 en log
    ln_D_fit = log_model((r_data, T_data), *popt)
    ss_res = np.sum((ln_D_obs - ln_D_fit)**2)
    ss_tot = np.sum((ln_D_obs - np.mean(ln_D_obs))**2)
    R2_log = 1 - ss_res / ss_tot

    # 7) Tracé optionnel
    if plot:
        plt.figure(figsize=(8, 6))
        plt.title("Fit : D = A * r^(-n) * (1 + c·T) * ...")

        unique_T = np.unique(T_data)
        import matplotlib.cm as cm
        colors = cm.viridis(np.linspace(0, 1, len(unique_T)))

        for T_val, col in zip(unique_T, colors):
            mask = (T_data == T_val)
            r_sub = r_data[mask]
            D_sub = D_data[mask]

            plt.loglog(r_sub, D_sub, 'o', color=col, label=f"T={T_val} (data)")

            r_min, r_max = r_sub.min(), r_sub.max()
            r_fit = np.logspace(np.log10(r_min), np.log10(r_max), 200)
            # D_fit
            D_fit = (A_opt * r_fit**(-n_opt)
                     * (1 + c_opt*T_val)
                     * np.exp(-mu_air_opt*r_fit)
                     * np.exp(-mu_pb_opt*T_val))

            plt.loglog(r_fit, D_fit, '-', color=col, label=f"T={T_val} (fit)")

        plt.xlabel("Distance r")
        plt.ylabel("Dose D")
        plt.legend()
        plt.grid(True, which="both", ls="--")

        formula_text = (
            f"D = {A_opt:.3g} * r^(-{n_opt:.3g}) * (1 + {c_opt:.3g}·T) "
            f"* exp(-{mu_air_opt:.3g}·r) * exp(-{mu_pb_opt:.3g}·T)\n"
            f"R² (log) = {R2_log:.4f}"
        )
        plt.text(0.05, 0.05, formula_text,
                 transform=plt.gca().transAxes,
                 bbox=dict(facecolor='white', alpha=0.7),
                 fontsize=10)
        plt.show()

    return A_opt, n_opt, c_opt, mu_air_opt, mu_pb_opt, R2_log, uncertainties
```

```python
import numpy as np
import matplotlib.pyplot as plt

def fit_polynomial_logr_thickness(r_data, T_data, D_data, plot=True):
    """
    Ajuste le modèle polynomial suivant (dans l'espace ln(D)) :
    
        ln(D) = b0
                + b1 * ln(r)
                + b2 * T
                + b3 * [ln(r)]^2
                + b4 * T^2
                + b5 * ln(r)*T
    
    Par régression linéaire multiple (moindres carrés).

    Paramètres
    ----------
    r_data : array-like
        Distances (m).
    T_data : array-like
        Épaisseurs (cm, m, peu importe tant que c'est cohérent dans D).
    D_data : array-like
        Doses (ou intensités) > 0.
    plot : bool
        Si True, trace un graphique log-log comparant les données et le fit.

    Retour
    ------
    betas : array of float
        Tableau [b0, b1, b2, b3, b4, b5].
    R2_log : float
        Coefficient de détermination dans l'espace ln(D).
    model_func : function
        Une fonction Python model_func(r, T) -> D prédite.
        Elle est vectorisée : r et T peuvent être scalaires ou tableaux.
    """

    # 1) Filtrage des données invalides
    r_data = np.array(r_data, dtype=float)
    T_data = np.array(T_data, dtype=float)
    D_data = np.array(D_data, dtype=float)

    valid_mask = (~np.isnan(D_data)) & (D_data > 0) & (r_data > 0)
    r_data = r_data[valid_mask]
    T_data = T_data[valid_mask]
    D_data = D_data[valid_mask]

    if len(D_data) == 0:
        raise ValueError("Aucune donnée valide pour l'ajustement.")

    # 2) Passage en log(D)
    ln_D = np.log(D_data)
    ln_r = np.log(r_data)

    # 3) Construction de la matrice X et du vecteur Y
    #    On veut : ln(D) = b0 + b1*ln(r) + b2*T + b3*(ln(r))^2 + b4*T^2 + b5*(ln(r)*T)
    X = []
    for i in range(len(r_data)):
        u = ln_r[i]
        t = T_data[i]
        X.append([1.0, u, t, u**2, t**2, u*t])
    X = np.array(X)  # shape (N, 6)

    # Y = ln(D)
    Y = ln_D

    # 4) Régression linéaire (lstsq)
    betas, residuals, rank, s = np.linalg.lstsq(X, Y, rcond=None)
    # betas = [b0, b1, b2, b3, b4, b5]

    # 5) Calcul du R^2 (dans l'espace ln(D))
    ln_D_fit = X.dot(betas)  # prédiction dans l'espace log
    ss_res = np.sum((ln_D - ln_D_fit)**2)
    ss_tot = np.sum((ln_D - np.mean(ln_D))**2)
    R2_log = 1 - ss_res / ss_tot

    # 6) Définition d'une fonction prédictive vectorisée
    def model_func(r, T):
        """
        Calcule la dose prédite D(r,T) = exp( b0 + b1 ln(r) + b2 T
                                             + b3 (ln(r))^2 + b4 T^2 + b5 ln(r)*T )
        """
        # Convertir r et T en tableaux pour profiter du broadcasting NumPy
        r = np.array(r, ndmin=1)
        T = np.array(T, ndmin=1)

        b0, b1, b2, b3, b4, b5 = betas

        ln_r = np.log(r)
        lnD = (b0
               + b1*ln_r
               + b2*T
               + b3*(ln_r**2)
               + b4*(T**2)
               + b5*(ln_r*T))
        return np.exp(lnD)

    # 7) Tracé optionnel
    if plot:
        plt.figure(figsize=(8, 6))
        plt.title("Fit polynomial en ln(r), T (2ᵉ ordre) sur ln(D)")
        # Pour distinguer chaque épaisseur
        unique_T = np.unique(T_data)

        import matplotlib.cm as cm
        colors = cm.viridis(np.linspace(0, 1, len(unique_T)))

        for t_val, col in zip(unique_T, colors):
            mask = (T_data == t_val)
            r_sub = r_data[mask]
            D_sub = D_data[mask]

            # Données expérimentales
            plt.loglog(r_sub, D_sub, "o", color=col, label=f"T={t_val} (data)")

            # Courbe ajustée
            if len(r_sub) > 0:
                r_min, r_max = r_sub.min(), r_sub.max()
                if r_min == r_max:
                    # Si on a qu'un seul point => pas de tracé lisse
                    continue
                r_fit = np.logspace(np.log10(r_min), np.log10(r_max), 200)
                D_fit = model_func(r_fit, t_val)
                plt.loglog(r_fit, D_fit, "-", color=col, label=f"T={t_val} (fit)")

        plt.xlabel("Distance r")
        plt.ylabel("Dose D")
        plt.grid(True, which="both", ls="--")
        plt.legend()

        formula_text = (
            "ln(D) = b0 + b1 ln(r) + b2 T + b3[ln(r)]^2 + b4 T^2 + b5 ln(r)*T\n"
            f"R²(log) = {R2_log:.4f}"
        )
        plt.text(0.05, 0.05, formula_text, transform=plt.gca().transAxes,
                 bbox=dict(facecolor='white', alpha=0.7), fontsize=10)
        plt.show()

    # 8) Retour : on renvoie les betas, le R2, et la fonction prédictive
    return betas, R2_log, model_func
```

```python
import numpy as np

# Exemple de données (x en mfp, y correspond au facteur de build-up mesuré, y_err l'incertitude)
x_data = np.array([0.5, 1.0, 1.5, 2.0, 2.5])
y_data = np.array([1.1, 1.3, 1.6, 1.9, 2.2])
y_err  = np.array([0.05, 0.05, 0.06, 0.07, 0.07])
```

```python
def model(x, alpha, beta, mu=1.0):
    """
    Modèle du facteur de build-up.
    x   : épaisseur exprimée en mfp (ou une autre unité adaptée)
    alpha, beta : paramètres du modèle
    mu  : facteur de conversion (si connu). Par défaut mu = 1.
    """
    return 1 + alpha * (mu * x)**beta

def chi2(alpha, beta):
    """
    Fonction chi2 qui compare le modèle aux données.
    """
    y_model = model(x_data, alpha, beta, mu=1.0)  # vous pouvez ajuster 'mu' ici si besoin
    return np.sum(((y_data - y_model) / y_err)**2)
```

```python
from iminuit import Minuit

# Initialisation des paramètres (exemple : alpha=1.0, beta=1.0)
m = Minuit(chi2, alpha=1.0, beta=1.0)

# Lancer la minimisation
m.migrad()

# Afficher les résultats
print("Valeurs optimales :")
print("alpha =", m.values["alpha"])
print("beta  =", m.values["beta"])

# Pour avoir une estimation des incertitudes:
print("Erreurs :")
print("d(alpha) =", m.errors["alpha"])
print("d(beta)  =", m.errors["beta"])
```

```python
import numpy as np
import matplotlib.pyplot as plt
from iminuit import Minuit
from iminuit.cost import LeastSquares
from ipywidgets import interact

# Pour la reproductibilité
rng = np.random.default_rng(1)

# Coefficient d'atténuation de l'air (en m^-1) pour une énergie donnée (exemple : 22 MeV)
mu_air = 0.002

# Définition du modèle dose en fonction de la distance r :
# D(r) = (C / r^2) * [1 + a*(mu_air*r)^b] * exp(-mu_air*r)
def dose_model(r, C, a, b, mu=mu_air):
    return (C / r**2) * (1 + a * (mu * r)**b) * np.exp(-mu * r)

# Génération de données synthétiques :
# Par exemple, nous générons des points pour r allant de 10 m à 500 m.
r_data = np.linspace(10, 500, 50)

# Paramètres "vrais" (exemple)
truth = (1e6, 2.0, 1.0)  # C = 1e6, a = 2.0, b = 1.0
y_true = dose_model(r_data, *truth, mu=mu_air)

# Bruit ajouté aux données (écart-type)
noise = 1e5
y_data = rng.normal(y_true, noise)
y_err  = noise * np.ones_like(y_data)

# Définition de la fonction de coût LeastSquares
# On fixe ici mu = mu_air dans le modèle
cost = LeastSquares(r_data, y_data, y_err,
                    lambda r, C, a, b: dose_model(r, C, a, b, mu=mu_air))

# Création de l'objet Minuit avec des valeurs initiales pour C, a et b
m = Minuit(cost, C=0.9e6, a=1.5, b=1.0)

# Première minimisation
m.migrad()

print("Résultats initiaux du fit :")
print("C =", m.values["C"])
print("a =", m.values["a"])
print("b =", m.values["b"])

# Lancement de l'interface interactive d'iminuit
# Cette commande ouvre des sliders pour ajuster les paramètres C, a et b en temps réel.
m.interactive()

# Pour visualiser l'ajustement, définissons une fonction de tracé
def plot_fit(C, a, b):
    plt.figure(figsize=(8,5))
    plt.errorbar(r_data, y_data, yerr=y_err, fmt='o', label="Données")
    r_fit = np.linspace(r_data.min(), r_data.max(), 300)
    y_fit = dose_model(r_fit, C, a, b, mu=mu_air)
    plt.plot(r_fit, y_fit, label=f"Fit: C={C:.0f}, a={a:.2f}, b={b:.2f}")
    plt.xlabel("Distance r (m)")
    plt.ylabel("Dose (unités arbitraires)")
    plt.legend()
    plt.yscale('log')  # Échelle logarithmique pour mieux visualiser la décroissance
    plt.show()

# Utilisation de ipywidgets.interact pour tracer en fonction des paramètres
interact(plot_fit, C=(0.5e6, 1.5e6, 1e4), a=(0.0, 5.0, 0.1), b=(0.0, 2.0, 0.05));
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
import matplotlib.cm as cm

def fit_data_to_distance_thickness(xdata, thickness_data, ydata, model=1, plot=True):
    """
    Ajuste différents modèles de la forme (en travaillant en log‑espace) :

    Modèle 1:
      y = A^(p) * x^(-k) * exp(-b*x) * exp(-c*thickness)
      Transformation logarithmique :
         ln(y) = p * ln(A) - k * ln(x) - b * x - c * thickness
      Paramètres à estimer : lnA, p, k, b, c

    Modèle 2:
      y = A^(a*thickness+b) * x^(-k) * exp(-B*x) * exp(-c*thickness)
      Transformation logarithmique :
         ln(y) = (a*thickness + b) * ln(A) - k * ln(x) - B * x - c * thickness
      Paramètres à estimer : lnA, a, b, k, B, c

    Modèle 3:
      y = A^(alpha1*thickness+beta1) * x^(-(alpha2*thickness+beta2))
      Transformation logarithmique :
         ln(y) = (alpha1*thickness + beta1) * ln(A) - (alpha2*thickness + beta2) * ln(x)
      Paramètres à estimer : lnA, alpha1, beta1, alpha2, beta2

    Renvoie :
      - popt         : le vecteur des paramètres optimaux (d'ordre fixé ci‑dessus)
      - R_squared    : le coefficient de détermination (calculé sur ln(y))
      - uncertainties: les écarts‑types sur les paramètres

    Le paramètre "model" permet de choisir :
      model = 1  => Modèle 1
      model = 2  => Modèle 2
      model = 3  => Modèle 3
    """
    # --- Préparation et filtrage des données ---
    xdata = np.array(xdata)
    thickness_data = np.array(thickness_data)
    ydata = np.array(ydata, dtype=float)

    # Conserver uniquement les y valides (strictement positifs et non NaN)
    valid_mask = ~np.isnan(ydata) & (ydata > 0)
    xdata = xdata[valid_mask]
    thickness_data = thickness_data[valid_mask]
    ydata = ydata[valid_mask]

    if len(ydata) == 0:
        raise ValueError("Pas de données valides pour l'ajustement.")

    ln_ydata = np.log(ydata)
    ln_x = np.log(xdata)

    # --- Choix du modèle et estimation initiale ---
    if model == 1:
        # Modèle 1 : y = A^(p) * x^(-k) * exp(-b*x) * exp(-c*thickness)
        # ln(y) = p * ln(A) - k * ln(x) - b*x - c*thickness
        # Pour obtenir une estimation initiale, on fixe p = 1 et on effectue une régression linéaire
        ones = np.ones_like(xdata)
        X = np.column_stack((ones, ln_x, xdata, thickness_data))
        Beta, residuals, rank, s = np.linalg.lstsq(X, ln_ydata, rcond=None)
        lnA_init, minus_k_init, minus_b_init, minus_c_init = Beta
        k_init = -minus_k_init
        b_init = -minus_b_init
        c_init = -minus_c_init
        p_init = 1.0  # On fixe p=1 dans l'initialisation

        initial_guess = [lnA_init, p_init, k_init, b_init, c_init]

        def log_model(vars, lnA, p, k, b, c):
            x, th = vars
            return p * lnA - k * np.log(x) - b * x - c * th

    elif model == 2:
        # Modèle 2 : y = A^(a*thickness+b) * x^(-k) * exp(-B*x) * exp(-c*thickness)
        # ln(y) = (a*thickness + b)*ln(A) - k*ln(x) - B*x - c*thickness
        # On fixe initialement a = 0 et b = 1 (pour obtenir une dépendance quasi nulle en thickness dans A)
        ones = np.ones_like(xdata)
        X = np.column_stack((ones, ln_x, xdata, thickness_data))
        Beta, residuals, rank, s = np.linalg.lstsq(X, ln_ydata, rcond=None)
        lnA_init, minus_k_init, minus_B_init, minus_c_init = Beta
        k_init = -minus_k_init
        B_init = -minus_B_init
        c_init = -minus_c_init
        a_init = 0.0
        b_init = 1.0

        initial_guess = [lnA_init, a_init, b_init, k_init, B_init, c_init]

        def log_model(vars, lnA, a, b, k, B, c):
            x, th = vars
            return (a * th + b) * lnA - k * np.log(x) - B * x - c * th

    elif model == 3:
        # Modèle 3 : y = A^(alpha1*thickness+beta1) * x^(-(alpha2*thickness+beta2))
        # ln(y) = (alpha1*thickness+beta1)*ln(A) - (alpha2*thickness+beta2)*ln(x)
        # Pour l'initialisation, on néglige la dépendance en thickness :
        #    on effectue une régression linéaire simple : ln(y) = ln(A) - beta2*ln(x)
        ones = np.ones_like(xdata)
        X_simple = np.column_stack((ones, ln_x))
        Beta_simple, residuals, rank, s = np.linalg.lstsq(X_simple, ln_ydata, rcond=None)
        lnA_init_simple, beta_term = Beta_simple
        # On fixe alpha1 = 0, alpha2 = 0 et on choisit beta1 = 1, beta2 = -beta_term
        alpha1_init = 0.0
        beta1_init = 1.0
        alpha2_init = 0.0
        beta2_init = -beta_term

        initial_guess = [lnA_init_simple, alpha1_init, beta1_init, alpha2_init, beta2_init]

        def log_model(vars, lnA, alpha1, beta1, alpha2, beta2):
            x, th = vars
            return (alpha1 * th + beta1) * lnA - (alpha2 * th + beta2) * np.log(x)
    else:
        raise ValueError("Modèle inconnu. Choisir model=1, 2 ou 3.")

    # --- Ajustement non linéaire via curve_fit ---
    popt, pcov = curve_fit(
        log_model,
        (xdata, thickness_data),
        ln_ydata,
        p0=initial_guess,
        maxfev=10000
    )
    perr = np.sqrt(np.diag(pcov))  # incertitudes sur les paramètres

    # Calcul du R² (sur ln(y))
    ln_yfit = log_model((xdata, thickness_data), *popt)
    ss_res = np.sum((ln_ydata - ln_yfit) ** 2)
    ss_tot = np.sum((ln_ydata - np.mean(ln_ydata)) ** 2)
    R_squared = 1 - ss_res / ss_tot

    # --- Tracé graphique (optionnel) ---
    if plot:
        plt.figure(figsize=(8, 6))
        plt.title(f"Fit des données (modèle {model})")

        # On trace pour chaque valeur unique de thickness
        unique_thicknesses = np.unique(thickness_data)
        colors = cm.viridis(np.linspace(0, 1, len(unique_thicknesses)))

        for th_val, col in zip(unique_thicknesses, colors):
            mask = (thickness_data == th_val)
            x_sub = xdata[mask]
            y_sub = ydata[mask]

            # Tracer les données expérimentales
            plt.loglog(x_sub, y_sub, 'o', color=col, label=f"ép={th_val}mm (data)")

            # Définir une gamme continue pour x
            x_fit = np.logspace(np.log10(x_sub.min()), np.log10(x_sub.max()), 100)
            # Calcul de y_fit selon le modèle
            if model == 1:
                lnA_opt, p_opt, k_opt, b_opt, c_opt = popt
                A_opt = np.exp(lnA_opt)
                y_fit = (A_opt**p_opt) * (x_fit**(-k_opt)) * np.exp(-b_opt * x_fit) * np.exp(-c_opt * th_val)
            elif model == 2:
                lnA_opt, a_opt, b_opt, k_opt, B_opt, c_opt = popt
                A_opt = np.exp(lnA_opt)
                y_fit = (A_opt**(a_opt * th_val + b_opt)) * (x_fit**(-k_opt)) * np.exp(-B_opt * x_fit) * np.exp(-c_opt * th_val)
            elif model == 3:
                lnA_opt, alpha1_opt, beta1_opt, alpha2_opt, beta2_opt = popt
                A_opt = np.exp(lnA_opt)
                y_fit = (A_opt**(alpha1_opt * th_val + beta1_opt)) * (x_fit** (-(alpha2_opt * th_val + beta2_opt)))

            plt.loglog(x_fit, y_fit, '-', color=col, label=f"ép={th_val}mm (fit)")

        plt.xlabel("Distance (m)")
        plt.ylabel("Dose (Gy)")
        plt.legend()
        plt.grid(True, which="both", ls="--")

        # Affichage d'une boîte texte avec la formule et le R²
        formula_text = f"Modèle {model} :\n"
        if model == 1:
            formula_text += ("y = A^(p) * x^(-k) * exp(-b*x) * exp(-c*thickness)\n" +
                             f"A = {np.exp(popt[0]):.3g}, p = {popt[1]:.3g}, k = {popt[2]:.3g}, " +
                             f"b = {popt[3]:.3g}, c = {popt[4]:.3g}\n")
        elif model == 2:
            formula_text += ("y = A^(a*thickness+b) * x^(-k) * exp(-B*x) * exp(-c*thickness)\n" +
                             f"A = {np.exp(popt[0]):.3g}, a = {popt[1]:.3g}, b = {popt[2]:.3g}, " +
                             f"k = {popt[3]:.3g}, B = {popt[4]:.3g}, c = {popt[5]:.3g}\n")
        elif model == 3:
            formula_text += ("y = A^(alpha1*thickness+beta1) * x^(-(alpha2*thickness+beta2))\n" +
                             f"A = {np.exp(popt[0]):.3g}, alpha1 = {popt[1]:.3g}, beta1 = {popt[2]:.3g}, " +
                             f"alpha2 = {popt[3]:.3g}, beta2 = {popt[4]:.3g}\n")
        formula_text += f"R² (log-space) = {R_squared:.4f}"
        plt.text(0.05, 0.05, formula_text, transform=plt.gca().transAxes,
                 bbox=dict(facecolor='white', alpha=0.8), fontsize=10)
        plt.show()

    return popt, R_squared, perr

# Imaginons que vous ayez un DataFrame 'df' avec les colonnes :
# "distance" (r), "thickness" (T), "dose" (D)
x = filtered_data["Distance (m)"].values
thickness = filtered_data["Thickness (cm)"].values
y_noisy = filtered_data["Dose"].values

popt, R2, uncertainties = fit_data_to_distance_thickness(x, thickness, y_noisy, model=1, plot=True)
print("Modèle 1 :")
print("Paramètres optimaux :", popt)
print("R² =", R2)
print("Incertitudes :", uncertainties)
# Pour tester les autres modèles, il suffit d'appeler la fonction avec model=2 ou model=3.
# Par exemple :
# popt2, R2_2, uncert2 = fit_data_to_distance_thickness(x, thickness, y_noisy, model=2, plot=True)
# popt3, R2_3, uncert3 = fit_data_to_distance_thickness(x, thickness, y_noisy, model=3, plot=True)
```

```python
import numpy as np
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt
import matplotlib.cm as cm

# ========================================================================
# MODÈLE 5 : Taylor Build-up Factor (forme exponentielle)
# ========================================================================
def fit_taylor_buildup(r_data, T_data, D_data, plot=True):
    """
    Modèle de build-up de Taylor :
    D(r,T) = (A/r²) × exp(-μ_air×r) × B(μ_pb×T) × exp(-μ_pb×T)
    
    où B(x) = 1 + α×x×exp(β×x)  (facteur de build-up de Taylor)
    
    Ce modèle capture mieux l'augmentation non-linéaire de la dose 
    due aux diffusions multiples dans le matériau.
    """
    r_data = np.array(r_data, dtype=float)
    T_data = np.array(T_data, dtype=float)
    D_data = np.array(D_data, dtype=float)
    
    valid = (~np.isnan(D_data)) & (D_data > 0) & (r_data > 0)
    r_data, T_data, D_data = r_data[valid], T_data[valid], D_data[valid]
    
    if len(D_data) == 0:
        raise ValueError("Aucune donnée valide")
    
    ln_D = np.log(D_data)
    
    # Estimation initiale
    ln_r = np.log(r_data)
    Y = ln_D + 2*ln_r
    X = np.column_stack([np.ones_like(r_data), r_data, T_data])
    Beta, _, _, _ = np.linalg.lstsq(X, Y, rcond=None)
    
    A_init = np.exp(Beta[0])
    mu_air_init = -Beta[1]
    mu_pb_init = -Beta[2]
    alpha_init = 0.5
    beta_init = -0.1
    
    initial_guess = [A_init, mu_air_init, mu_pb_init, alpha_init, beta_init]
    
    def log_model(vars, A, mu_air, mu_pb, alpha, beta):
        r, T = vars
        x = mu_pb * T
        buildup = 1 + alpha * x * np.exp(beta * x)
        return (np.log(A) - 2*np.log(r) - mu_air*r 
                + np.log(buildup) - mu_pb*T)
    
    try:
        popt, pcov = curve_fit(log_model, (r_data, T_data), ln_D,
                              p0=initial_guess, maxfev=50000,
                              bounds=([0, 0, 0, -10, -5], 
                                     [np.inf, 1, 1, 10, 5]))
        
        A_opt, mu_air_opt, mu_pb_opt, alpha_opt, beta_opt = popt
        perr = np.sqrt(np.diag(pcov))
        
        ln_D_fit = log_model((r_data, T_data), *popt)
        ss_res = np.sum((ln_D - ln_D_fit)**2)
        ss_tot = np.sum((ln_D - np.mean(ln_D))**2)
        R2 = 1 - ss_res/ss_tot
        
        if plot:
            plt.figure(figsize=(8, 6))
            plt.title("Modèle Taylor Build-up")
            unique_T = np.unique(T_data)
            colors = cm.viridis(np.linspace(0, 1, len(unique_T)))
            
            for T_val, col in zip(unique_T, colors):
                mask = (T_data == T_val)
                r_sub, D_sub = r_data[mask], D_data[mask]
                plt.loglog(r_sub, D_sub, 'o', color=col, label=f"T={T_val} cm (data)")
                
                r_fit = np.logspace(np.log10(r_sub.min()), np.log10(r_sub.max()), 200)
                x_fit = mu_pb_opt * T_val
                buildup_fit = 1 + alpha_opt * x_fit * np.exp(beta_opt * x_fit)
                D_fit = (A_opt/r_fit**2 * np.exp(-mu_air_opt*r_fit) 
                        * buildup_fit * np.exp(-mu_pb_opt*T_val))
                plt.loglog(r_fit, D_fit, '-', color=col, label=f"T={T_val} cm (fit)")
            
            plt.xlabel("Distance r (m)")
            plt.ylabel("Dose D (Gy)")
            plt.legend(fontsize=8)
            plt.grid(True, which="both", ls="--")
            
            formula_text = (f"B(x) = 1 + {alpha_opt:.3g}·x·exp({beta_opt:.3g}·x)\n"
                          f"R² = {R2:.6f}")
            plt.text(0.05, 0.05, formula_text, transform=plt.gca().transAxes,
                    bbox=dict(facecolor='white', alpha=0.7), fontsize=9)
            plt.show()
        
        return A_opt, mu_air_opt, mu_pb_opt, alpha_opt, beta_opt, R2, list(perr)
    
    except Exception as e:
        print(f"Erreur Taylor: {e}")
        return None, None, None, None, None, 0, [0]*5


# ========================================================================
# MODÈLE 6 : Berger Build-up (polynomiale)
# ========================================================================
def fit_berger_buildup(r_data, T_data, D_data, plot=True):
    """
    Modèle de build-up polynomial de Berger :
    B(x) = 1 + c₁×x + c₂×x² 
    
    Plus adapté pour des épaisseurs moyennes (1-20 MFP)
    """
    r_data = np.array(r_data, dtype=float)
    T_data = np.array(T_data, dtype=float)
    D_data = np.array(D_data, dtype=float)
    
    valid = (~np.isnan(D_data)) & (D_data > 0) & (r_data > 0)
    r_data, T_data, D_data = r_data[valid], T_data[valid], D_data[valid]
    
    if len(D_data) == 0:
        raise ValueError("Aucune donnée valide")
    
    ln_D = np.log(D_data)
    ln_r = np.log(r_data)
    
    # Estimation initiale simplifiée
    Y = ln_D + 2*ln_r
    X = np.column_stack([np.ones_like(r_data), r_data, T_data])
    Beta, _, _, _ = np.linalg.lstsq(X, Y, rcond=None)
    
    A_init = np.exp(Beta[0])
    mu_air_init = max(0.001, -Beta[1])
    mu_pb_init = max(0.01, -Beta[2])
    
    initial_guess = [A_init, mu_air_init, mu_pb_init, 0.1, 0.01]
    
    def log_model(vars, A, mu_air, mu_pb, c1, c2):
        r, T = vars
        x = mu_pb * T
        buildup = 1 + c1*x + c2*x**2
        # Éviter log de nombres négatifs
        buildup = np.maximum(buildup, 0.01)
        return (np.log(A) - 2*np.log(r) - mu_air*r 
                + np.log(buildup) - mu_pb*T)
    
    try:
        popt, pcov = curve_fit(log_model, (r_data, T_data), ln_D,
                              p0=initial_guess, maxfev=50000,
                              bounds=([0, 0, 0, -5, -1], 
                                     [np.inf, 1, 1, 5, 1]))
        
        A_opt, mu_air_opt, mu_pb_opt, c1_opt, c2_opt = popt
        perr = np.sqrt(np.diag(pcov))
        
        ln_D_fit = log_model((r_data, T_data), *popt)
        ss_res = np.sum((ln_D - ln_D_fit)**2)
        ss_tot = np.sum((ln_D - np.mean(ln_D))**2)
        R2 = 1 - ss_res/ss_tot
        
        if plot:
            plt.figure(figsize=(8, 6))
            plt.title("Modèle Berger Build-up (polynomial)")
            unique_T = np.unique(T_data)
            colors = cm.viridis(np.linspace(0, 1, len(unique_T)))
            
            for T_val, col in zip(unique_T, colors):
                mask = (T_data == T_val)
                r_sub, D_sub = r_data[mask], D_data[mask]
                plt.loglog(r_sub, D_sub, 'o', color=col, label=f"T={T_val} cm (data)")
                
                r_fit = np.logspace(np.log10(r_sub.min()), np.log10(r_sub.max()), 200)
                x_fit = mu_pb_opt * T_val
                buildup_fit = 1 + c1_opt*x_fit + c2_opt*x_fit**2
                D_fit = (A_opt/r_fit**2 * np.exp(-mu_air_opt*r_fit) 
                        * buildup_fit * np.exp(-mu_pb_opt*T_val))
                plt.loglog(r_fit, D_fit, '-', color=col, label=f"T={T_val} cm (fit)")
            
            plt.xlabel("Distance r (m)")
            plt.ylabel("Dose D (Gy)")
            plt.legend(fontsize=8)
            plt.grid(True, which="both", ls="--")
            
            formula_text = (f"B(x) = 1 + {c1_opt:.3g}·x + {c2_opt:.3g}·x²\n"
                          f"R² = {R2:.6f}")
            plt.text(0.05, 0.05, formula_text, transform=plt.gca().transAxes,
                    bbox=dict(facecolor='white', alpha=0.7), fontsize=9)
            plt.show()
        
        return A_opt, mu_air_opt, mu_pb_opt, c1_opt, c2_opt, R2, list(perr)
    
    except Exception as e:
        print(f"Erreur Berger: {e}")
        return None, None, None, None, None, 0, [0]*5


# ========================================================================
# MODÈLE 7 : GP (Geometric Progression) Build-up
# ========================================================================
def fit_gp_buildup(r_data, T_data, D_data, plot=True):
    """
    Modèle GP (Geometric Progression) - ANSI/ANS-6.4.3 :
    B(x) = 1 + (b-1) × [K^x - 1] / [K - 1]
    
    où K = c×x^α + d×tanh(x/X - 2) - tanh(-2)
    
    C'est le modèle le plus utilisé dans les standards de radioprotection
    """
    r_data = np.array(r_data, dtype=float)
    T_data = np.array(T_data, dtype=float)
    D_data = np.array(D_data, dtype=float)
    
    valid = (~np.isnan(D_data)) & (D_data > 0) & (r_data > 0)
    r_data, T_data, D_data = r_data[valid], T_data[valid], D_data[valid]
    
    if len(D_data) == 0:
        raise ValueError("Aucune donnée valide")
    
    ln_D = np.log(D_data)
    ln_r = np.log(r_data)
    
    # Estimation initiale
    Y = ln_D + 2*ln_r
    X = np.column_stack([np.ones_like(r_data), r_data, T_data])
    Beta, _, _, _ = np.linalg.lstsq(X, Y, rcond=None)
    
    A_init = np.exp(Beta[0])
    mu_air_init = max(0.001, -Beta[1])
    mu_pb_init = max(0.01, -Beta[2])
    
    # Paramètres GP : b, c, alpha, d, X_k
    initial_guess = [A_init, mu_air_init, mu_pb_init, 1.5, 1.0, 0.5, 0.1, 2.0]
    
    def log_model(vars, A, mu_air, mu_pb, b, c, alpha, d, X_k):
        r, T = vars
        x = mu_pb * T
        
        # Éviter division par zéro et valeurs invalides
        x = np.maximum(x, 1e-6)
        
        # Calcul de K avec protection
        K = c * x**alpha + d * (np.tanh(x/X_k - 2) - np.tanh(-2))
        K = np.maximum(K, 1.001)  # K doit être > 1
        
        # Build-up GP avec gestion vectorielle
        # Utiliser np.where au lieu de if/else pour gérer les tableaux
        K_diff = K - 1
        buildup = np.where(
            np.abs(K_diff) < 1e-6,
            1 + (b - 1) * x,  # Cas limite K ≈ 1
            1 + (b - 1) * (K**x - 1) / K_diff  # Cas général
        )
        
        buildup = np.maximum(buildup, 0.1)  # Protection
        
        return (np.log(A) - 2*np.log(r) - mu_air*r 
                + np.log(buildup) - mu_pb*T)
    
    try:
        popt, pcov = curve_fit(log_model, (r_data, T_data), ln_D,
                              p0=initial_guess, maxfev=100000,
                              bounds=([0, 0, 0, 1, 0, 0, -5, 0.1], 
                                     [np.inf, 1, 1, 10, 5, 2, 5, 10]))
        
        A_opt, mu_air_opt, mu_pb_opt, b_opt, c_opt, alpha_opt, d_opt, X_k_opt = popt
        perr = np.sqrt(np.diag(pcov))
        
        ln_D_fit = log_model((r_data, T_data), *popt)
        ss_res = np.sum((ln_D - ln_D_fit)**2)
        ss_tot = np.sum((ln_D - np.mean(ln_D))**2)
        R2 = 1 - ss_res/ss_tot
        
        if plot:
            plt.figure(figsize=(8, 6))
            plt.title("Modèle GP (ANSI/ANS-6.4.3) Build-up")
            unique_T = np.unique(T_data)
            colors = cm.viridis(np.linspace(0, 1, len(unique_T)))
            
            for T_val, col in zip(unique_T, colors):
                mask = (T_data == T_val)
                r_sub, D_sub = r_data[mask], D_data[mask]
                plt.loglog(r_sub, D_sub, 'o', color=col, label=f"T={T_val} cm (data)")
                
                r_fit = np.logspace(np.log10(r_sub.min()), np.log10(r_sub.max()), 200)
                x_fit = mu_pb_opt * T_val
                
                K_fit = c_opt * x_fit**alpha_opt + d_opt * (np.tanh(x_fit/X_k_opt - 2) - np.tanh(-2))
                K_fit = max(K_fit, 1.001)
                
                if np.abs(K_fit - 1) < 1e-6:
                    buildup_fit = 1 + (b_opt - 1) * x_fit
                else:
                    buildup_fit = 1 + (b_opt - 1) * (K_fit**x_fit - 1) / (K_fit - 1)
                
                D_fit = (A_opt/r_fit**2 * np.exp(-mu_air_opt*r_fit) 
                        * buildup_fit * np.exp(-mu_pb_opt*T_val))
                plt.loglog(r_fit, D_fit, '-', color=col, label=f"T={T_val} cm (fit)")
            
            plt.xlabel("Distance r (m)")
            plt.ylabel("Dose D (Gy)")
            plt.legend(fontsize=8)
            plt.grid(True, which="both", ls="--")
            
            formula_text = (f"GP: b={b_opt:.3g}, c={c_opt:.3g}, α={alpha_opt:.3g}\n"
                          f"R² = {R2:.6f}")
            plt.text(0.05, 0.05, formula_text, transform=plt.gca().transAxes,
                    bbox=dict(facecolor='white', alpha=0.7), fontsize=9)
            plt.show()
        
        return A_opt, mu_air_opt, mu_pb_opt, b_opt, c_opt, alpha_opt, d_opt, X_k_opt, R2, list(perr)
    
    except Exception as e:
        import traceback
        print(f"Erreur GP: {e}")
        print(f"Traceback: {traceback.format_exc()}")
        return None, None, None, None, None, None, None, None, 0, [0]*8


print("✅ Modèles avancés définis : Taylor, Berger, GP")
```

```python
# ========================================================================
# MODÈLE 8 : Atténuation avec Skyshine (diffusion atmosphérique)
# ========================================================================
def fit_skyshine_model(r_data, T_data, D_data, plot=True):
    """
    Modèle incluant la contribution du rayonnement diffusé par l'atmosphère :
    D(r,T) = D_direct(r,T) + D_skyshine(r,T)
    
    D_direct = (A/r²) × exp(-μ_air×r) × exp(-μ_pb×T)
    D_skyshine = (B/r^n) × exp(-μ_sky×r) × exp(-μ_pb×T/2)
    
    Le skyshine a un exposant différent et une atténuation réduite dans l'écran
    """
    r_data = np.array(r_data, dtype=float)
    T_data = np.array(T_data, dtype=float)
    D_data = np.array(D_data, dtype=float)
    
    valid = (~np.isnan(D_data)) & (D_data > 0) & (r_data > 0)
    r_data, T_data, D_data = r_data[valid], T_data[valid], D_data[valid]
    
    if len(D_data) == 0:
        raise ValueError("Aucune donnée valide")
    
    ln_D = np.log(D_data)
    
    # Estimation initiale
    ln_r = np.log(r_data)
    Y = ln_D + 2*ln_r
    X = np.column_stack([np.ones_like(r_data), r_data, T_data])
    Beta, _, _, _ = np.linalg.lstsq(X, Y, rcond=None)
    
    A_init = np.exp(Beta[0]) * 0.9
    B_init = np.exp(Beta[0]) * 0.1
    mu_air_init = max(0.001, -Beta[1])
    mu_pb_init = max(0.01, -Beta[2])
    
    initial_guess = [A_init, B_init, mu_air_init, mu_air_init*0.5, mu_pb_init, 1.5]
    
    def log_model(vars, A, B, mu_air, mu_sky, mu_pb, n_sky):
        r, T = vars
        D_direct = (A/r**2) * np.exp(-mu_air*r) * np.exp(-mu_pb*T)
        D_skyshine = (B/r**n_sky) * np.exp(-mu_sky*r) * np.exp(-mu_pb*T*0.5)
        D_total = D_direct + D_skyshine
        D_total = np.maximum(D_total, 1e-10)
        return np.log(D_total)
    
    try:
        popt, pcov = curve_fit(log_model, (r_data, T_data), ln_D,
                              p0=initial_guess, maxfev=50000,
                              bounds=([0, 0, 0, 0, 0, 0.5], 
                                     [np.inf, np.inf, 1, 1, 1, 3]))
        
        A_opt, B_opt, mu_air_opt, mu_sky_opt, mu_pb_opt, n_sky_opt = popt
        perr = np.sqrt(np.diag(pcov))
        
        ln_D_fit = log_model((r_data, T_data), *popt)
        ss_res = np.sum((ln_D - ln_D_fit)**2)
        ss_tot = np.sum((ln_D - np.mean(ln_D))**2)
        R2 = 1 - ss_res/ss_tot
        
        if plot:
            plt.figure(figsize=(8, 6))
            plt.title("Modèle avec Skyshine (diffusion atmosphérique)")
            unique_T = np.unique(T_data)
            colors = cm.viridis(np.linspace(0, 1, len(unique_T)))
            
            for T_val, col in zip(unique_T, colors):
                mask = (T_data == T_val)
                r_sub, D_sub = r_data[mask], D_data[mask]
                plt.loglog(r_sub, D_sub, 'o', color=col, label=f"T={T_val} cm (data)")
                
                r_fit = np.logspace(np.log10(r_sub.min()), np.log10(r_sub.max()), 200)
                D_direct_fit = (A_opt/r_fit**2) * np.exp(-mu_air_opt*r_fit) * np.exp(-mu_pb_opt*T_val)
                D_skyshine_fit = (B_opt/r_fit**n_sky_opt) * np.exp(-mu_sky_opt*r_fit) * np.exp(-mu_pb_opt*T_val*0.5)
                D_total_fit = D_direct_fit + D_skyshine_fit
                
                plt.loglog(r_fit, D_total_fit, '-', color=col, label=f"T={T_val} cm (fit)")
            
            plt.xlabel("Distance r (m)")
            plt.ylabel("Dose D (Gy)")
            plt.legend(fontsize=8)
            plt.grid(True, which="both", ls="--")
            
            formula_text = (f"Skyshine: B={B_opt:.3e}, n={n_sky_opt:.3g}\n"
                          f"R² = {R2:.6f}")
            plt.text(0.05, 0.05, formula_text, transform=plt.gca().transAxes,
                    bbox=dict(facecolor='white', alpha=0.7), fontsize=9)
            plt.show()
        
        return A_opt, B_opt, mu_air_opt, mu_sky_opt, mu_pb_opt, n_sky_opt, R2, list(perr)
    
    except Exception as e:
        print(f"Erreur Skyshine: {e}")
        return None, None, None, None, None, None, 0, [0]*6


# ========================================================================
# MODÈLE 9 : Modèle hybride avec terme de correction
# ========================================================================
def fit_hybrid_model(r_data, T_data, D_data, plot=True):
    """
    Modèle hybride combinant :
    - Loi de puissance variable avec la distance
    - Build-up polynomial dans l'écran
    - Terme de correction exponentiel pour grandes distances
    
    D(r,T) = A × r^(-k) × (1 + c₁×T + c₂×T²) × exp(-μ_air×r - μ_pb×T) × [1 + ε×exp(-r/λ)]
    
    Le terme [1 + ε×exp(-r/λ)] corrige les écarts à courte distance
    """
    r_data = np.array(r_data, dtype=float)
    T_data = np.array(T_data, dtype=float)
    D_data = np.array(D_data, dtype=float)
    
    valid = (~np.isnan(D_data)) & (D_data > 0) & (r_data > 0)
    r_data, T_data, D_data = r_data[valid], T_data[valid], D_data[valid]
    
    if len(D_data) == 0:
        raise ValueError("Aucune donnée valide")
    
    ln_D = np.log(D_data)
    
    # Estimation initiale simplifiée
    ln_r = np.log(r_data)
    Y = ln_D + 2*ln_r
    X = np.column_stack([np.ones_like(r_data), r_data, T_data, T_data**2])
    Beta, _, _, _ = np.linalg.lstsq(X, Y, rcond=None)
    
    A_init = np.exp(Beta[0])
    mu_air_init = max(0.001, -Beta[1])
    mu_pb_init = max(0.01, -Beta[2])
    
    initial_guess = [A_init, 1.8, mu_air_init, mu_pb_init, 0.1, 0.01, 0.1, 10]
    
    def log_model(vars, A, k, mu_air, mu_pb, c1, c2, epsilon, lambda_r):
        r, T = vars
        buildup = 1 + c1*T + c2*T**2
        buildup = np.maximum(buildup, 0.1)
        correction = 1 + epsilon * np.exp(-r/lambda_r)
        
        return (np.log(A) - k*np.log(r) + np.log(buildup) 
                - mu_air*r - mu_pb*T + np.log(correction))
    
    try:
        popt, pcov = curve_fit(log_model, (r_data, T_data), ln_D,
                              p0=initial_guess, maxfev=50000,
                              bounds=([0, 0.5, 0, 0, -2, -0.5, -1, 1], 
                                     [np.inf, 3, 1, 1, 2, 0.5, 1, 1000]))
        
        A_opt, k_opt, mu_air_opt, mu_pb_opt, c1_opt, c2_opt, eps_opt, lambda_opt = popt
        perr = np.sqrt(np.diag(pcov))
        
        ln_D_fit = log_model((r_data, T_data), *popt)
        ss_res = np.sum((ln_D - ln_D_fit)**2)
        ss_tot = np.sum((ln_D - np.mean(ln_D))**2)
        R2 = 1 - ss_res/ss_tot
        
        if plot:
            plt.figure(figsize=(8, 6))
            plt.title("Modèle Hybride avec correction")
            unique_T = np.unique(T_data)
            colors = cm.viridis(np.linspace(0, 1, len(unique_T)))
            
            for T_val, col in zip(unique_T, colors):
                mask = (T_data == T_val)
                r_sub, D_sub = r_data[mask], D_data[mask]
                plt.loglog(r_sub, D_sub, 'o', color=col, label=f"T={T_val} cm (data)")
                
                r_fit = np.logspace(np.log10(r_sub.min()), np.log10(r_sub.max()), 200)
                buildup_fit = 1 + c1_opt*T_val + c2_opt*T_val**2
                correction_fit = 1 + eps_opt * np.exp(-r_fit/lambda_opt)
                D_fit = (A_opt * r_fit**(-k_opt) * buildup_fit 
                        * np.exp(-mu_air_opt*r_fit - mu_pb_opt*T_val) * correction_fit)
                plt.loglog(r_fit, D_fit, '-', color=col, label=f"T={T_val} cm (fit)")
            
            plt.xlabel("Distance r (m)")
            plt.ylabel("Dose D (Gy)")
            plt.legend(fontsize=8)
            plt.grid(True, which="both", ls="--")
            
            formula_text = (f"k={k_opt:.3g}, ε={eps_opt:.3g}, λ={lambda_opt:.3g}\n"
                          f"R² = {R2:.6f}")
            plt.text(0.05, 0.05, formula_text, transform=plt.gca().transAxes,
                    bbox=dict(facecolor='white', alpha=0.7), fontsize=9)
            plt.show()
        
        return A_opt, k_opt, mu_air_opt, mu_pb_opt, c1_opt, c2_opt, eps_opt, lambda_opt, R2, list(perr)
    
    except Exception as e:
        print(f"Erreur Hybride: {e}")
        return None, None, None, None, None, None, None, None, 0, [0]*8


print("✅ Modèles supplémentaires définis : Skyshine, Hybride")
```

# 🧪 Nouvelles variantes de modèles

Exploration de modèles alternatifs basés sur :
- Atténuation multi-exponentielle
- Build-up avec dépendance énergétique
- Modèles empiriques de la littérature (Chilton, Capo)
- Combinaisons hybrides optimisées

```python
import numpy as np
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt
import matplotlib.cm as cm

# ========================================================================
# MODÈLE 10 : Double exponentielle (multi-composantes)
# ========================================================================
def fit_double_exponential(r_data, T_data, D_data, plot=True):
    """
    Modèle à double exponentielle pour capturer deux régimes :
    - Courte distance : atténuation rapide (composante primaire)
    - Longue distance : atténuation lente (diffusion)
    
    D(r,T) = A₁×r^(-k₁)×exp(-μ₁×r)×exp(-μ_pb×T) + 
             A₂×r^(-k₂)×exp(-μ₂×r)×exp(-μ_pb×T/2)
    """
    r_data = np.array(r_data, dtype=float)
    T_data = np.array(T_data, dtype=float)
    D_data = np.array(D_data, dtype=float)
    
    valid = (~np.isnan(D_data)) & (D_data > 0) & (r_data > 0)
    r_data, T_data, D_data = r_data[valid], T_data[valid], D_data[valid]
    
    if len(D_data) == 0:
        raise ValueError("Aucune donnée valide")
    
    # Estimation initiale
    A1_init = 1.0
    A2_init = 0.1
    k1_init, k2_init = 2.0, 1.0
    mu1_init, mu2_init = 0.03, 0.005
    mu_pb_init = 0.1
    
    initial_guess = [A1_init, k1_init, mu1_init, A2_init, k2_init, mu2_init, mu_pb_init]
    
    def model(vars, A1, k1, mu1, A2, k2, mu2, mu_pb):
        r, T = vars
        comp1 = A1 * r**(-k1) * np.exp(-mu1 * r) * np.exp(-mu_pb * T)
        comp2 = A2 * r**(-k2) * np.exp(-mu2 * r) * np.exp(-mu_pb * T / 2)
        return comp1 + comp2
    
    try:
        popt, pcov = curve_fit(model, (r_data, T_data), D_data,
                              p0=initial_guess, maxfev=50000,
                              bounds=([0, 0.5, 0, 0, 0.5, 0, 0], 
                                     [10, 3, 0.5, 10, 3, 0.1, 1]))
        
        A1, k1, mu1, A2, k2, mu2, mu_pb = popt
        perr = np.sqrt(np.diag(pcov))
        
        D_fit = model((r_data, T_data), *popt)
        ss_res = np.sum((D_data - D_fit)**2)
        ss_tot = np.sum((D_data - np.mean(D_data))**2)
        R2 = 1 - ss_res/ss_tot
        
        if plot:
            plt.figure(figsize=(10, 6))
            plt.title("Modèle Double Exponentielle")
            unique_T = np.unique(T_data)
            colors = cm.viridis(np.linspace(0, 1, len(unique_T)))
            
            for T_val, col in zip(unique_T, colors):
                mask = (T_data == T_val)
                r_sub, D_sub = r_data[mask], D_data[mask]
                plt.loglog(r_sub, D_sub, 'o', color=col, label=f"T={T_val} cm (data)", markersize=6)
                
                r_fit = np.logspace(np.log10(r_sub.min()), np.log10(r_sub.max()), 200)
                D_total = model((r_fit, np.full_like(r_fit, T_val)), *popt)
                plt.loglog(r_fit, D_total, '-', color=col, linewidth=2)
                
                # Composantes séparées
                comp1 = A1 * r_fit**(-k1) * np.exp(-mu1 * r_fit) * np.exp(-mu_pb * T_val)
                comp2 = A2 * r_fit**(-k2) * np.exp(-mu2 * r_fit) * np.exp(-mu_pb * T_val / 2)
                plt.loglog(r_fit, comp1, '--', color=col, alpha=0.3, linewidth=1)
                plt.loglog(r_fit, comp2, ':', color=col, alpha=0.3, linewidth=1)
            
            plt.xlabel("Distance r (m)", fontsize=12)
            plt.ylabel("Dose D (Gy)", fontsize=12)
            plt.legend(fontsize=8)
            plt.grid(True, which="both", ls="--", alpha=0.3)
            
            ratio = A2/A1
            formula_text = (f"R² = {R2:.6f}\n"
                          f"A₂/A₁ = {ratio:.3f}\n"
                          f"μ₁ = {mu1:.4f}, μ₂ = {mu2:.4f}")
            plt.text(0.05, 0.05, formula_text, transform=plt.gca().transAxes,
                    bbox=dict(facecolor='white', alpha=0.8), fontsize=10)
            plt.tight_layout()
            plt.show()
        
        return A1, k1, mu1, A2, k2, mu2, mu_pb, R2, list(perr)
    
    except Exception as e:
        print(f"Erreur Double Expo: {e}")
        return None, None, None, None, None, None, None, 0, [0]*7


# ========================================================================
# MODÈLE 11 : Chilton Build-up (formule empirique classique)
# ========================================================================
def fit_chilton_buildup(r_data, T_data, D_data, plot=True):
    """
    Formule de Chilton pour le facteur de build-up :
    B(μ×x) = A×e^(-α₁×μ×x) - (A-1)×e^(-α₂×μ×x)
    
    Où A et α sont des paramètres empiriques dépendant du matériau
    """
    r_data = np.array(r_data, dtype=float)
    T_data = np.array(T_data, dtype=float)
    D_data = np.array(D_data, dtype=float)
    
    valid = (~np.isnan(D_data)) & (D_data > 0) & (r_data > 0)
    r_data, T_data, D_data = r_data[valid], T_data[valid], D_data[valid]
    
    if len(D_data) == 0:
        raise ValueError("Aucune donnée valide")
    
    ln_D = np.log(D_data)
    
    # Estimation initiale
    ln_r = np.log(r_data)
    Y = ln_D + 2*ln_r
    X = np.column_stack([np.ones_like(r_data), r_data, T_data])
    Beta, _, _, _ = np.linalg.lstsq(X, Y, rcond=None)
    
    A0_init = np.exp(Beta[0])
    mu_air_init = max(0.001, -Beta[1])
    mu_pb_init = max(0.01, -Beta[2])
    A_chilton_init = 2.0
    alpha1_init = 1.0
    alpha2_init = 0.5
    
    initial_guess = [A0_init, mu_air_init, mu_pb_init, A_chilton_init, alpha1_init, alpha2_init]
    
    def log_model(vars, A0, mu_air, mu_pb, A_ch, alpha1, alpha2):
        r, T = vars
        x = mu_pb * T
        
        # Build-up de Chilton
        buildup = A_ch * np.exp(-alpha1 * x) - (A_ch - 1) * np.exp(-alpha2 * x)
        buildup = np.maximum(buildup, 0.1)
        
        return (np.log(A0) - 2*np.log(r) - mu_air*r 
                + np.log(buildup) - mu_pb*T)
    
    try:
        popt, pcov = curve_fit(log_model, (r_data, T_data), ln_D,
                              p0=initial_guess, maxfev=50000,
                              bounds=([0, 0, 0, 1, 0, 0], 
                                     [np.inf, 1, 1, 10, 5, 5]))
        
        A0, mu_air, mu_pb, A_ch, alpha1, alpha2 = popt
        perr = np.sqrt(np.diag(pcov))
        
        ln_D_fit = log_model((r_data, T_data), *popt)
        ss_res = np.sum((ln_D - ln_D_fit)**2)
        ss_tot = np.sum((ln_D - np.mean(ln_D))**2)
        R2 = 1 - ss_res/ss_tot
        
        if plot:
            plt.figure(figsize=(8, 6))
            plt.title("Modèle Chilton Build-up")
            unique_T = np.unique(T_data)
            colors = cm.viridis(np.linspace(0, 1, len(unique_T)))
            
            for T_val, col in zip(unique_T, colors):
                mask = (T_data == T_val)
                r_sub, D_sub = r_data[mask], D_data[mask]
                plt.loglog(r_sub, D_sub, 'o', color=col, label=f"T={T_val} cm (data)")
                
                r_fit = np.logspace(np.log10(r_sub.min()), np.log10(r_sub.max()), 200)
                x_fit = mu_pb * T_val
                buildup_fit = A_ch * np.exp(-alpha1 * x_fit) - (A_ch - 1) * np.exp(-alpha2 * x_fit)
                D_fit = (A0/r_fit**2 * np.exp(-mu_air*r_fit) 
                        * buildup_fit * np.exp(-mu_pb*T_val))
                plt.loglog(r_fit, D_fit, '-', color=col)
            
            plt.xlabel("Distance r (m)")
            plt.ylabel("Dose D (Gy)")
            plt.legend(fontsize=8)
            plt.grid(True, which="both", ls="--")
            
            formula_text = (f"Chilton: A={A_ch:.3g}, α₁={alpha1:.3g}, α₂={alpha2:.3g}\n"
                          f"R² = {R2:.6f}")
            plt.text(0.05, 0.05, formula_text, transform=plt.gca().transAxes,
                    bbox=dict(facecolor='white', alpha=0.7), fontsize=9)
            plt.show()
        
        return A0, mu_air, mu_pb, A_ch, alpha1, alpha2, R2, list(perr)
    
    except Exception as e:
        print(f"Erreur Chilton: {e}")
        return None, None, None, None, None, None, 0, [0]*6


# ========================================================================
# MODÈLE 12 : Capo Build-up (amélioration de Chilton)
# ========================================================================  
def fit_capo_buildup(r_data, T_data, D_data, plot=True):
    """
    Formule de Capo (amélioration de Chilton) :
    B(x) = 1 + (b-1) × [x + c×x²] / [1 + d×x]
    
    Plus précis pour les épaisseurs intermédiaires
    """
    r_data = np.array(r_data, dtype=float)
    T_data = np.array(T_data, dtype=float)
    D_data = np.array(D_data, dtype=float)
    
    valid = (~np.isnan(D_data)) & (D_data > 0) & (r_data > 0)
    r_data, T_data, D_data = r_data[valid], T_data[valid], D_data[valid]
    
    if len(D_data) == 0:
        raise ValueError("Aucune donnée valide")
    
    ln_D = np.log(D_data)
    
    # Estimation initiale
    ln_r = np.log(r_data)
    Y = ln_D + 2*ln_r
    X = np.column_stack([np.ones_like(r_data), r_data, T_data])
    Beta, _, _, _ = np.linalg.lstsq(X, Y, rcond=None)
    
    A_init = np.exp(Beta[0])
    mu_air_init = max(0.001, -Beta[1])
    mu_pb_init = max(0.01, -Beta[2])
    b_init, c_init, d_init = 2.0, 0.1, 0.1
    
    initial_guess = [A_init, mu_air_init, mu_pb_init, b_init, c_init, d_init]
    
    def log_model(vars, A, mu_air, mu_pb, b, c, d):
        r, T = vars
        x = mu_pb * T
        
        # Build-up de Capo
        numerator = x + c * x**2
        denominator = 1 + d * x
        buildup = 1 + (b - 1) * numerator / np.maximum(denominator, 0.1)
        buildup = np.maximum(buildup, 0.1)
        
        return (np.log(A) - 2*np.log(r) - mu_air*r 
                + np.log(buildup) - mu_pb*T)
    
    try:
        popt, pcov = curve_fit(log_model, (r_data, T_data), ln_D,
                              p0=initial_guess, maxfev=50000,
                              bounds=([0, 0, 0, 1, -1, 0], 
                                     [np.inf, 1, 1, 10, 2, 5]))
        
        A, mu_air, mu_pb, b, c, d = popt
        perr = np.sqrt(np.diag(pcov))
        
        ln_D_fit = log_model((r_data, T_data), *popt)
        ss_res = np.sum((ln_D - ln_D_fit)**2)
        ss_tot = np.sum((ln_D - np.mean(ln_D))**2)
        R2 = 1 - ss_res/ss_tot
        
        if plot:
            plt.figure(figsize=(8, 6))
            plt.title("Modèle Capo Build-up")
            unique_T = np.unique(T_data)
            colors = cm.viridis(np.linspace(0, 1, len(unique_T)))
            
            for T_val, col in zip(unique_T, colors):
                mask = (T_data == T_val)
                r_sub, D_sub = r_data[mask], D_data[mask]
                plt.loglog(r_sub, D_sub, 'o', color=col, label=f"T={T_val} cm (data)")
                
                r_fit = np.logspace(np.log10(r_sub.min()), np.log10(r_sub.max()), 200)
                x_fit = mu_pb * T_val
                numerator = x_fit + c * x_fit**2
                denominator = 1 + d * x_fit
                buildup_fit = 1 + (b - 1) * numerator / max(denominator, 0.1)
                D_fit = (A/r_fit**2 * np.exp(-mu_air*r_fit) 
                        * buildup_fit * np.exp(-mu_pb*T_val))
                plt.loglog(r_fit, D_fit, '-', color=col)
            
            plt.xlabel("Distance r (m)")
            plt.ylabel("Dose D (Gy)")
            plt.legend(fontsize=8)
            plt.grid(True, which="both", ls="--")
            
            formula_text = (f"Capo: b={b:.3g}, c={c:.3g}, d={d:.3g}\n"
                          f"R² = {R2:.6f}")
            plt.text(0.05, 0.05, formula_text, transform=plt.gca().transAxes,
                    bbox=dict(facecolor='white', alpha=0.7), fontsize=9)
            plt.show()
        
        return A, mu_air, mu_pb, b, c, d, R2, list(perr)
    
    except Exception as e:
        print(f"Erreur Capo: {e}")
        return None, None, None, None, None, None, 0, [0]*6


print("✅ Nouvelles variantes définies : Double Expo, Chilton, Capo")
```

```python
# ========================================================================
# MODÈLE 13 : Skyshine optimisé avec build-up dans l'écran
# ========================================================================
def fit_skyshine_enhanced(r_data, T_data, D_data, plot=True):
    """
    Version améliorée du Skyshine avec build-up polynomial dans l'écran :
    D_direct = (A/r²) × exp(-μ_air×r) × B_screen(T) × exp(-μ_pb×T)
    D_skyshine = (B/r^n) × exp(-μ_sky×r) × exp(-μ_pb×T/f_atten)
    
    où B_screen(T) = 1 + c₁×T + c₂×T²
    """
    r_data = np.array(r_data, dtype=float)
    T_data = np.array(T_data, dtype=float)
    D_data = np.array(D_data, dtype=float)
    
    valid = (~np.isnan(D_data)) & (D_data > 0) & (r_data > 0)
    r_data, T_data, D_data = r_data[valid], T_data[valid], D_data[valid]
    
    if len(D_data) == 0:
        raise ValueError("Aucune donnée valide")
    
    ln_D = np.log(D_data)
    
    # Estimation initiale basée sur le skyshine simple
    A_init = 1.0
    B_init = 0.1
    mu_air_init = 0.03
    mu_sky_init = 0.005
    mu_pb_init = 0.1
    n_sky_init = 1.2
    c1_init, c2_init = 0.0, 0.0
    f_atten_init = 2.0
    
    initial_guess = [A_init, B_init, mu_air_init, mu_sky_init, mu_pb_init, 
                     n_sky_init, c1_init, c2_init, f_atten_init]
    
    def log_model(vars, A, B, mu_air, mu_sky, mu_pb, n_sky, c1, c2, f_atten):
        r, T = vars
        
        # Build-up dans l'écran
        buildup_screen = 1 + c1*T + c2*T**2
        buildup_screen = np.maximum(buildup_screen, 0.1)
        
        # Composante directe avec build-up
        D_direct = (A/r**2 * np.exp(-mu_air*r) 
                   * buildup_screen * np.exp(-mu_pb*T))
        
        # Composante skyshine avec atténuation réduite dans l'écran
        D_skyshine = (B * r**(-n_sky) * np.exp(-mu_sky*r) 
                     * np.exp(-mu_pb*T/f_atten))
        
        D_total = D_direct + D_skyshine
        return np.log(D_total)
    
    try:
        popt, pcov = curve_fit(log_model, (r_data, T_data), ln_D,
                              p0=initial_guess, maxfev=100000,
                              bounds=([0, 0, 0, 0, 0, 0.5, -1, -0.5, 1], 
                                     [10, 1, 0.1, 0.05, 1, 2, 1, 0.5, 10]))
        
        A, B, mu_air, mu_sky, mu_pb, n_sky, c1, c2, f_atten = popt
        perr = np.sqrt(np.diag(pcov))
        
        ln_D_fit = log_model((r_data, T_data), *popt)
        ss_res = np.sum((ln_D - ln_D_fit)**2)
        ss_tot = np.sum((ln_D - np.mean(ln_D))**2)
        R2 = 1 - ss_res/ss_tot
        
        if plot:
            fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
            
            # Plot 1 : Fit global
            unique_T = np.unique(T_data)
            colors = cm.viridis(np.linspace(0, 1, len(unique_T)))
            
            for T_val, col in zip(unique_T, colors):
                mask = (T_data == T_val)
                r_sub, D_sub = r_data[mask], D_data[mask]
                ax1.loglog(r_sub, D_sub, 'o', color=col, 
                          label=f"T={T_val} cm", markersize=6)
                
                r_fit = np.logspace(np.log10(r_sub.min()), np.log10(r_sub.max()), 200)
                D_fit = np.exp(log_model((r_fit, np.full_like(r_fit, T_val)), *popt))
                ax1.loglog(r_fit, D_fit, '-', color=col, linewidth=2)
            
            ax1.set_xlabel("Distance r (m)", fontsize=11)
            ax1.set_ylabel("Dose D (Gy)", fontsize=11)
            ax1.set_title(f"Skyshine Enhanced - R² = {R2:.6f}", fontsize=12)
            ax1.legend(fontsize=8)
            ax1.grid(True, which="both", ls="--", alpha=0.3)
            
            # Plot 2 : Build-up dans l'écran
            T_range = np.linspace(0, 20, 100)
            buildup_curve = 1 + c1*T_range + c2*T_range**2
            ax2.plot(T_range, buildup_curve, 'b-', linewidth=2, label='Build-up screen')
            ax2.axhline(y=1, color='k', linestyle='--', alpha=0.3, label='Sans build-up')
            ax2.scatter(T_data, 1 + c1*T_data + c2*T_data**2, 
                       c='red', s=30, alpha=0.5, zorder=5)
            ax2.set_xlabel("Épaisseur écran T (cm)", fontsize=11)
            ax2.set_ylabel("Facteur de build-up", fontsize=11)
            ax2.set_title(f"Build-up : B = 1 + {c1:.3g}·T + {c2:.3g}·T²", fontsize=11)
            ax2.legend(fontsize=9)
            ax2.grid(True, alpha=0.3)
            
            plt.tight_layout()
            plt.show()
        
        return A, B, mu_air, mu_sky, mu_pb, n_sky, c1, c2, f_atten, R2, list(perr)
    
    except Exception as e:
        print(f"Erreur Skyshine Enhanced: {e}")
        return None, None, None, None, None, None, None, None, None, 0, [0]*9


# ========================================================================
# MODÈLE 14 : Super Hybride (combinaison optimale)
# ========================================================================
def fit_super_hybrid(r_data, T_data, D_data, plot=True):
    """
    Combinaison ultime :
    - Exposant variable avec distance ET épaisseur
    - Build-up polynomial dans l'écran
    - Correction exponentielle à longue distance
    - Terme de couplage distance-épaisseur
    
    D = A × r^(-k(T)) × B(T) × exp(-μ_air×r) × exp(-μ_pb×T) × C(r)
    
    où :
    k(T) = k₀ + k₁×T
    B(T) = 1 + b₁×T + b₂×T²
    C(r) = 1 + ε×exp(-r/λ)
    """
    r_data = np.array(r_data, dtype=float)
    T_data = np.array(T_data, dtype=float)
    D_data = np.array(D_data, dtype=float)
    
    valid = (~np.isnan(D_data)) & (D_data > 0) & (r_data > 0)
    r_data, T_data, D_data = r_data[valid], T_data[valid], D_data[valid]
    
    if len(D_data) == 0:
        raise ValueError("Aucune donnée valide")
    
    ln_D = np.log(D_data)
    
    # Estimation initiale sophistiquée
    ln_r = np.log(r_data)
    Y = ln_D + 2*ln_r
    X = np.column_stack([np.ones_like(r_data), r_data, T_data, r_data*T_data])
    Beta, _, _, _ = np.linalg.lstsq(X, Y, rcond=None)
    
    A_init = np.exp(Beta[0])
    mu_air_init = max(0.001, -Beta[1])
    mu_pb_init = max(0.01, -Beta[2])
    
    k0_init, k1_init = 1.8, -0.01
    b1_init, b2_init = 0.0, 0.0
    eps_init, lambda_init = -0.5, 100
    
    initial_guess = [A_init, k0_init, k1_init, mu_air_init, mu_pb_init, 
                     b1_init, b2_init, eps_init, lambda_init]
    
    def log_model(vars, A, k0, k1, mu_air, mu_pb, b1, b2, eps, lambda_r):
        r, T = vars
        
        # Exposant variable
        k_eff = k0 + k1*T
        k_eff = np.clip(k_eff, 0.5, 3)
        
        # Build-up dans l'écran
        buildup = 1 + b1*T + b2*T**2
        buildup = np.maximum(buildup, 0.1)
        
        # Correction longue distance
        correction = 1 + eps * np.exp(-r/lambda_r)
        
        return (np.log(A) - k_eff*np.log(r) + np.log(buildup)
                - mu_air*r - mu_pb*T + np.log(np.abs(correction)))
    
    try:
        popt, pcov = curve_fit(log_model, (r_data, T_data), ln_D,
                              p0=initial_guess, maxfev=100000,
                              bounds=([0, 0.5, -0.1, 0, 0, -2, -0.5, -5, 10], 
                                     [np.inf, 3, 0.1, 0.5, 1, 2, 0.5, 5, 2000]))
        
        A, k0, k1, mu_air, mu_pb, b1, b2, eps, lambda_r = popt
        perr = np.sqrt(np.diag(pcov))
        
        ln_D_fit = log_model((r_data, T_data), *popt)
        ss_res = np.sum((ln_D - ln_D_fit)**2)
        ss_tot = np.sum((ln_D - np.mean(ln_D))**2)
        R2 = 1 - ss_res/ss_tot
        
        if plot:
            plt.figure(figsize=(10, 6))
            plt.title(f"Super Hybride - R² = {R2:.6f}")
            unique_T = np.unique(T_data)
            colors = cm.viridis(np.linspace(0, 1, len(unique_T)))
            
            for T_val, col in zip(unique_T, colors):
                mask = (T_data == T_val)
                r_sub, D_sub = r_data[mask], D_data[mask]
                plt.loglog(r_sub, D_sub, 'o', color=col, 
                          label=f"T={T_val} cm", markersize=7)
                
                r_fit = np.logspace(np.log10(r_sub.min()), np.log10(r_sub.max()), 200)
                D_fit = np.exp(log_model((r_fit, np.full_like(r_fit, T_val)), *popt))
                plt.loglog(r_fit, D_fit, '-', color=col, linewidth=2.5)
            
            plt.xlabel("Distance r (m)", fontsize=12)
            plt.ylabel("Dose D (Gy)", fontsize=12)
            plt.legend(fontsize=9)
            plt.grid(True, which="both", ls="--", alpha=0.3)
            
            info_text = (f"k(T) = {k0:.3g} + {k1:.3g}×T\n"
                        f"B(T) = 1 + {b1:.3g}×T + {b2:.3g}×T²\n"
                        f"ε = {eps:.3g}, λ = {lambda_r:.1f} m")
            plt.text(0.05, 0.05, info_text, transform=plt.gca().transAxes,
                    bbox=dict(facecolor='white', alpha=0.85), fontsize=10)
            plt.tight_layout()
            plt.show()
        
        return A, k0, k1, mu_air, mu_pb, b1, b2, eps, lambda_r, R2, list(perr)
    
    except Exception as e:
        print(f"Erreur Super Hybrid: {e}")
        return None, None, None, None, None, None, None, None, None, 0, [0]*9


print("✅ Modèles ultra-optimisés définis : Skyshine Enhanced, Super Hybride")
```

## 🔵 Modèles Standards (1-4)

### Modèle 1 : Exposant Variable
$$
D(r, T) = A \cdot r^{-(a \cdot T + b)} \cdot e^{-\mu_{\text{air}} \cdot r} \cdot e^{-\mu_{\text{wall}} \cdot T}
$$
- **Paramètres** : 5 (A, a, b, μ_air, μ_wall)
- **Principe** : L'exposant géométrique varie linéairement avec l'épaisseur

### Modèle 2 : Build-up Linéaire (1/r²)
$$
D(r, T) = A \cdot \frac{1}{r^2} \cdot e^{-\mu_{\text{air}} \cdot r} \cdot B(T) \cdot e^{-\mu_{\text{wall}} \cdot T}
$$
avec $B(T) = 1 + b \cdot T$ (build-up linéaire)
- **Paramètres** : 4 (A, b, μ_air, μ_wall)
- **Principe** : Loi inverse carré avec build-up linéaire

### Modèle 3 : Build-up avec Exposant Ajustable
$$
D(r, T) = A \cdot r^{-n} \cdot e^{-\mu_{\text{air}} \cdot r} \cdot B(T) \cdot e^{-\mu_{\text{wall}} \cdot T}
$$
avec $B(T) = 1 + b \cdot T$
- **Paramètres** : 5 (A, n, b, μ_air, μ_wall)
- **Principe** : Exposant géométrique libre + build-up linéaire

### Modèle 4 : Polynomial 2D en Log-espace
$$
\ln(D) = a_0 + a_1 \ln(r) + a_2 \ln(r)^2 + b_1 T + b_2 T^2 + c \ln(r) \cdot T
$$
- **Paramètres** : 6 (a₀, a₁, a₂, b₁, b₂, c)
- **Principe** : Ajustement polynomial sans contrainte physique

---

## 🟢 Modèles Physiques Avancés (5-9)

### Modèle 5 : Taylor Build-up
$$
D(r, T) = A \cdot \frac{1}{r^2} \cdot e^{-\mu_{\text{air}} \cdot r} \cdot B_{\text{Taylor}}(T) \cdot e^{-\mu_{\text{wall}} \cdot T}
$$
avec $B_{\text{Taylor}}(T) = 1 + K \cdot T \cdot e^{-\alpha T}$
- **Paramètres** : 5 (A, K, α, μ_air, μ_wall)
- **Principe** : Formule de Taylor pour le build-up

### Modèle 6 : Berger Build-up (Polynomial)
$$
D(r, T) = A \cdot \frac{1}{r^2} \cdot e^{-\mu_{\text{air}} \cdot r} \cdot B_{\text{Berger}}(T) \cdot e^{-\mu_{\text{wall}} \cdot T}
$$
avec $B_{\text{Berger}}(T) = 1 + c_1 T + c_2 T^2$
- **Paramètres** : 5 (A, c₁, c₂, μ_air, μ_wall)
- **Principe** : Build-up polynomial de Berger

### Modèle 7 : GP Build-up (ANSI/ANS-6.4.3)
$$
D(r, T) = A \cdot \frac{1}{r^2} \cdot e^{-\mu_{\text{air}} \cdot r} \cdot B_{\text{GP}}(T) \cdot e^{-\mu_{\text{wall}} \cdot T}
$$
avec $B_{\text{GP}}(T) = 1 + \frac{(K-1)(K \mu T - 1)}{K-1 + \tanh(\alpha T)}$
- **Paramètres** : 8 (A, K, α, μ_air, μ_wall, + paramètres GP)
- **Principe** : Geometric Progression (norme ANSI)

### Modèle 8 : Skyshine (Diffusion Atmosphérique)
$$
D(r, T) = \underbrace{A \cdot \frac{1}{r^2} \cdot e^{-\mu_{\text{air}} \cdot r} \cdot e^{-\mu_{\text{wall}} \cdot T}}_{\text{Direct}} + \underbrace{B \cdot r^{-n} \cdot e^{-\mu_{\text{sky}} \cdot r} \cdot e^{-\mu_{\text{wall}} \cdot T}}_{\text{Skyshine}}
$$
- **Paramètres** : 6 (A, B, n, μ_air, μ_sky, μ_wall)
- **Principe** : Deux composantes (directe + diffusion atmosphérique)

### Modèle 9 : Hybride Corrigé
$$
D(r, T) = A \cdot r^{-n} \cdot e^{-\mu_{\text{air}} \cdot r} \cdot (1 + b \cdot T) \cdot e^{-\mu_{\text{wall}} \cdot T} \cdot (1 + \frac{c}{r^p})
$$
- **Paramètres** : 8 (A, n, b, c, p, μ_air, μ_wall)
- **Principe** : Build-up + correction longue distance

## 🟡 Nouvelles Variantes (10-12)

### Modèle 10 : Double Exponentielle
$$
D(r, T) = A \cdot \frac{1}{r^2} \cdot (c_1 e^{-\mu_1 r} + c_2 e^{-\mu_2 r}) \cdot e^{-\mu_{\text{wall}} \cdot T}
$$
- **Paramètres** : 7 (A, c₁, c₂, μ₁, μ₂, μ_wall)
- **Principe** : Deux composantes d'atténuation dans l'air

### Modèle 11 : Chilton Build-up
$$
D(r, T) = A \cdot \frac{1}{r^2} \cdot e^{-\mu_{\text{air}} \cdot r} \cdot B_{\text{Chilton}}(T) \cdot e^{-\mu_{\text{wall}} \cdot T}
$$
avec $B_{\text{Chilton}}(T) = A_0 e^{-\alpha_1 T} + (1-A_0)e^{-\alpha_2 T}$
- **Paramètres** : 6 (A, A₀, α₁, α₂, μ_air, μ_wall)
- **Principe** : Build-up de Chilton (double exponentielle)

### Modèle 12 : Capo Build-up
$$
D(r, T) = A \cdot \frac{1}{r^2} \cdot e^{-\mu_{\text{air}} \cdot r} \cdot B_{\text{Capo}}(T) \cdot e^{-\mu_{\text{wall}} \cdot T}
$$
avec $B_{\text{Capo}}(T) = 1 + C(\mu T)^{\beta}$
- **Paramètres** : 6 (A, C, β, μ, μ_air, μ_wall)
- **Principe** : Build-up de Capo (loi de puissance)

```python
# 📈 Comparaison visuelle de TOUS les modèles sur les mêmes données

import matplotlib.pyplot as plt
import matplotlib.cm as cm
import numpy as np

# Liste des modèles à tester avec leurs fonctions et nombre de paramètres
models_to_test = [
    ("Skyshine (6p)", fit_skyshine_model if 'fit_skyshine_model' in dir() else None, 6),
    ("Skyshine Enhanced (9p)", fit_skyshine_enhanced if 'fit_skyshine_enhanced' in dir() else None, 9),
    ("Super Hybride (9p)", fit_super_hybrid if 'fit_super_hybrid' in dir() else None, 9),
    ("Hybride corrigé (8p)", fit_hybrid_model if 'fit_hybrid_model' in dir() else None, 8),
    ("Double Exponentielle (7p)", fit_double_exponential if 'fit_double_exponential' in dir() else None, 7),
]

# Dictionnaire pour stocker les résultats
results_comparison = {}

print("="*80)
print(" 🔬 COMPARAISON DES MODÈLES SUR PHOTONS + ACIER + U(93.2)")
print("="*80)

# Fit de chaque modèle
for model_name, fit_func, n_params in models_to_test:
    if fit_func is None:
        print(f"\n⚠️  {model_name} : Fonction non disponible")
        continue
    
    try:
        print(f"\n🔄 Fit en cours : {model_name}...")
        result = fit_func(r_test, T_test, D_test, plot=False)
        
        if result is not None and result[0] is not None:
            # Les fonctions retournent des tuples, extraire R2 et calculer RMSE
            # Format général : (...params..., R2, errors)
            # R2 est l'avant-dernier élément
            params = result[:-2]  # Tous les paramètres
            R2 = result[-2]  # R2 est l'avant-dernier
            
            # Calculer RMSE en évaluant le modèle
            # On doit recalculer D_fit selon le type de modèle
            if "Skyshine Enhanced" in model_name:
                A, B, mu_air, mu_sky, mu_pb, n_sky, c1, c2, f_atten = params
                buildup = 1 + c1*T_test + c2*T_test**2
                buildup = np.maximum(buildup, 0.1)
                D_fit = (A/r_test**2 * np.exp(-mu_air*r_test) * buildup * np.exp(-mu_pb*T_test) +
                         B * r_test**(-n_sky) * np.exp(-mu_sky*r_test) * np.exp(-mu_pb*T_test/f_atten))
            elif "Skyshine" in model_name and "Enhanced" not in model_name:
                A, B, mu_air, mu_sky, mu_pb, n_sky = params
                D_fit = ((A/r_test**2) * np.exp(-mu_air*r_test) * np.exp(-mu_pb*T_test) +
                         (B/r_test**n_sky) * np.exp(-mu_sky*r_test) * np.exp(-mu_pb*T_test*0.5))
            elif "Super Hybride" in model_name:
                # Modèle 14 : r^(-(k0+k1*T)) avec build-up et correction distance
                A, k0, k1, mu_air, mu_pb, b1, b2, eps, lambda_r = params
                k_eff = k0 + k1*T_test
                buildup = 1 + b1*T_test + b2*T_test**2
                correction = 1 + eps * np.exp(-r_test / lambda_r)
                D_fit = A * r_test**(-k_eff) * np.exp(-mu_air*r_test) * buildup * np.exp(-mu_pb*T_test) * correction
            elif "Hybride corrigé" in model_name:
                # Modèle 9 : r^-k avec build-up polynomial et correction exponentielle
                A, k, mu_air, mu_pb, c1, c2, eps, lambda_r = params
                buildup = 1 + c1*T_test + c2*T_test**2
                correction = 1 + eps * np.exp(-r_test / lambda_r)
                D_fit = A * r_test**(-k) * np.exp(-mu_air*r_test) * buildup * np.exp(-mu_pb*T_test) * correction
            elif "Double Exponentielle" in model_name:
                # Modèle 10 : Deux composantes d'atténuation
                A1, k1, mu1, A2, k2, mu2, mu_pb = params
                comp1 = A1 * r_test**(-k1) * np.exp(-mu1 * r_test) * np.exp(-mu_pb * T_test)
                comp2 = A2 * r_test**(-k2) * np.exp(-mu2 * r_test) * np.exp(-mu_pb * T_test / 2)
                D_fit = comp1 + comp2
            else:
                D_fit = D_test  # Fallback
            
            RMSE = np.sqrt(np.mean((D_test - D_fit)**2))
            
            results_comparison[model_name] = {
                'params': params,
                'R2': R2,
                'RMSE': RMSE,
                'n_points': len(D_test)
            }
            print(f"   ✅ R² = {R2:.6f}, RMSE = {RMSE:.3e} Gy")
        else:
            print(f"   ❌ Échec du fit")
    except Exception as e:
        print(f"   ❌ Erreur : {str(e)[:100]}")
        import traceback
        traceback.print_exc()

print("\n" + "="*80)
print(f" ✅ {len(results_comparison)} modèles ont été fittés avec succès")
print("="*80)
```

```python
# 🎯 Fonction de visualisation interactive pour un modèle

def plot_model_comparison(model_name, results_dict, r_test, T_test, D_test, figsize=(12, 8)):
    """
    Visualise les résultats d'un modèle spécifique avec toutes les épaisseurs.
    
    Parameters:
    -----------
    model_name : str
        Nom du modèle à visualiser (doit être dans results_dict)
    results_dict : dict
        Dictionnaire contenant les résultats de tous les modèles
    r_test, T_test, D_test : array-like
        Données de test
    figsize : tuple
        Taille de la figure
    """
    
    if model_name not in results_dict:
        print(f"❌ Modèle '{model_name}' non trouvé dans les résultats.")
        print(f"📋 Modèles disponibles : {list(results_dict.keys())}")
        return
    
    # Extraire les résultats du modèle
    result = results_dict[model_name]
    params = result['params']
    R2 = result['R2']
    RMSE = result['RMSE']
    n_points = result['n_points']
    
    # Identifier les épaisseurs uniques
    unique_T = np.unique(T_test)
    
    # Palette de couleurs pour les épaisseurs
    colors = cm.viridis(np.linspace(0, 1, len(unique_T)))
    
    # Créer la figure
    fig, ax = plt.subplots(figsize=figsize)
    
    # Pour chaque épaisseur
    for T_val, color in zip(unique_T, colors):
        # Données expérimentales pour cette épaisseur
        mask = (T_test == T_val)
        r_subset = r_test[mask]
        D_subset = D_test[mask]
        
        # Tracer les données expérimentales
        ax.loglog(r_subset, D_subset, 'o', color=color, markersize=10, 
                 label=f'T={T_val} cm (données MCNP 6.1)', 
                 markeredgewidth=1.5, markeredgecolor='k', alpha=0.8, zorder=5)
        
        # Générer les prédictions sur une grille fine
        r_fit = np.logspace(np.log10(r_subset.min()), np.log10(r_subset.max()), 300)
        
        # Calculer les prédictions selon le modèle
        if 'Skyshine Enhanced' in model_name:
            A, B, mu_air, mu_sky, mu_pb, n_sky, c1, c2, f_atten = params
            buildup = np.maximum(1 + c1*T_val + c2*T_val**2, 0.1)
            D_direct = (A/r_fit**2 * np.exp(-mu_air*r_fit) * buildup * np.exp(-mu_pb*T_val))
            D_sky = (B * r_fit**(-n_sky) * np.exp(-mu_sky*r_fit) * np.exp(-mu_pb*T_val/f_atten))
            D_pred = D_direct + D_sky
            
        elif 'Skyshine' in model_name and 'Enhanced' not in model_name:
            A, B, mu_air, mu_sky, mu_pb, n_sky = params
            D_direct = A/r_fit**2 * np.exp(-mu_air*r_fit) * np.exp(-mu_pb*T_val)
            D_sky = B * r_fit**(-n_sky) * np.exp(-mu_sky*r_fit) * np.exp(-mu_pb*T_val*0.5)
            D_pred = D_direct + D_sky
            
        elif 'Super Hybride' in model_name:
            # Modèle 14 : r^(-(k0+k1*T)) avec build-up et correction distance
            A, k0, k1, mu_air, mu_pb, b1, b2, eps, lambda_r = params
            k_eff = k0 + k1*T_val
            buildup = 1 + b1*T_val + b2*T_val**2
            correction = 1 + eps * np.exp(-r_fit / lambda_r)
            D_pred = A * r_fit**(-k_eff) * np.exp(-mu_air*r_fit) * buildup * np.exp(-mu_pb*T_val) * correction
            
        elif 'Hybride corrigé' in model_name:
            # Modèle 9 : r^-k avec build-up polynomial et correction exponentielle
            A, k, mu_air, mu_pb, c1, c2, eps, lambda_r = params
            buildup = 1 + c1*T_val + c2*T_val**2
            correction = 1 + eps * np.exp(-r_fit / lambda_r)
            D_pred = A * r_fit**(-k) * np.exp(-mu_air*r_fit) * buildup * np.exp(-mu_pb*T_val) * correction
            
        elif 'Double Exponentielle' in model_name:
            # Modèle 10 : Deux composantes d'atténuation
            A1, k1, mu1, A2, k2, mu2, mu_pb = params
            comp1 = A1 * r_fit**(-k1) * np.exp(-mu1 * r_fit) * np.exp(-mu_pb * T_val)
            comp2 = A2 * r_fit**(-k2) * np.exp(-mu2 * r_fit) * np.exp(-mu_pb * T_val / 2)
            D_pred = comp1 + comp2
        else:
            print(f"⚠️  Type de modèle '{model_name}' non reconnu pour les prédictions")
            return
        
        # Tracer la prédiction
        ax.loglog(r_fit, D_pred, '-', color=color, linewidth=3, alpha=0.9, zorder=3)
    
    # Mise en forme du graphique
    ax.set_xlabel('Distance r (m)', fontsize=14, fontweight='bold')
    ax.set_ylabel('Dose D (Gy)', fontsize=14, fontweight='bold')
    ax.set_title(f'{model_name}\nR² = {R2:.6f} | RMSE = {RMSE:.2e} Gy | {n_points} points', 
                 fontsize=16, fontweight='bold', pad=20)
    ax.legend(fontsize=11, loc='best', framealpha=0.95, edgecolor='black', fancybox=True)
    ax.grid(True, which='both', ls='--', alpha=0.5, linewidth=0.8)
    ax.tick_params(labelsize=11)
    
    plt.tight_layout()
    plt.show()

print("✅ Fonction plot_model_comparison() définie")
```

## 🔍 Analyse de la contribution des paramètres

Identification des paramètres essentiels vs redondants dans les 2 meilleurs modèles

```python
# 🔍 ANALYSE DE SENSIBILITÉ DES PARAMÈTRES

print("="*100)
print(" 🔍 ANALYSE DE LA CONTRIBUTION DES PARAMÈTRES - TOP 2 MODÈLES")
print("="*100)

# ============================================================================
# MODÈLE #1 : SKYSHINE ENHANCED
# ============================================================================
print("\n" + "="*100)
print(" 🥇 MODÈLE #1 : SKYSHINE ENHANCED (9 paramètres)")
print("="*100)

# Récupérer les paramètres du champion
A_ref, B_ref, mu_air_ref, mu_sky_ref, mu_pb_ref, n_sky_ref, c1_ref, c2_ref, f_atten_ref, R2_ref, unc_ref = \
    fit_skyshine_enhanced(r_data, T_data, D_data, plot=False)

param_names = ['A', 'B', 'μ_air', 'μ_sky', 'μ_pb', 'n_sky', 'c₁', 'c₂', 'f_atten']
param_values = [A_ref, B_ref, mu_air_ref, mu_sky_ref, mu_pb_ref, n_sky_ref, c1_ref, c2_ref, f_atten_ref]
param_uncertainties = unc_ref

print(f"\n📊 RÉFÉRENCE : R² = {R2_ref:.6f}\n")

# Analyse 1 : Incertitude relative de chaque paramètre
print("📋 TABLEAU 1 : INCERTITUDES RELATIVES")
print("-" * 80)

sensitivity_data = []
for i, (name, val, unc) in enumerate(zip(param_names, param_values, param_uncertainties)):
    rel_unc = (unc / abs(val)) * 100 if abs(val) > 1e-10 else np.inf
    confidence = "✅ Bien déterminé" if rel_unc < 20 else "⚠️ Incertain" if rel_unc < 50 else "❌ Très incertain"
    
    sensitivity_data.append({
        'Paramètre': name,
        'Valeur': f"{val:.4e}" if abs(val) < 0.01 or abs(val) > 100 else f"{val:.4f}",
        'Incertitude': f"{unc:.4e}" if abs(unc) < 0.01 or abs(unc) > 100 else f"{unc:.4f}",
        'Incertitude rel. (%)': f"{rel_unc:.1f}",
        'Confiance': confidence
    })

df_sensitivity = pd.DataFrame(sensitivity_data)
print(df_sensitivity.to_string(index=False))

# Analyse 2 : Test de nullité (paramètre peut-il être fixé à 0 ?)
print("\n\n📋 TABLEAU 2 : TEST DE SIGNIFICATIVITÉ (peut-on fixer à 0 ?)")
print("-" * 80)

significance_tests = []
for i, (name, val, unc) in enumerate(zip(param_names, param_values, param_uncertainties)):
    # Test : |valeur| > 2×incertitude (95% de confiance)
    is_significant = abs(val) > 2 * unc
    z_score = abs(val) / unc if unc > 1e-10 else np.inf
    
    if name in ['A', 'B', 'μ_air', 'μ_sky', 'μ_pb', 'n_sky']:
        # Paramètres physiques : ne peuvent pas être nuls
        verdict = "🔒 ESSENTIEL (physique)"
        can_be_zero = "Non (physique)"
    elif is_significant and z_score > 3:
        verdict = "✅ SIGNIFICATIF"
        can_be_zero = "Non (>3σ)"
    elif is_significant:
        verdict = "⚠️ MARGINAL"
        can_be_zero = "Non (>2σ)"
    else:
        verdict = "❌ NON SIGNIFICATIF"
        can_be_zero = "Oui (<2σ)"
    
    significance_tests.append({
        'Paramètre': name,
        'Valeur': f"{val:.4f}" if abs(val) < 100 else f"{val:.2e}",
        '|Valeur|/Incertitude': f"{z_score:.2f}",
        'Peut être 0 ?': can_be_zero,
        'Verdict': verdict
    })

df_significance = pd.DataFrame(significance_tests)
print(df_significance.to_string(index=False))

# Analyse 3 : Test d'ablation (retrait de chaque paramètre)
print("\n\n📋 TABLEAU 3 : TEST D'ABLATION - Impact sur R²")
print("-" * 80)
print("(Fixer chaque paramètre à sa valeur triviale et mesurer la dégradation)")

ablation_results = []

# Test pour c₁ et c₂ (build-up) - peuvent être fixés à 0
for param_idx, param_name in [(6, 'c₁'), (7, 'c₂')]:
    # Créer une copie des paramètres avec celui-ci fixé
    params_test = param_values.copy()
    params_test[param_idx] = 0.0  # Fixer à 0
    
    # Évaluer R² avec ce paramètre fixé
    A_t, B_t, mu_air_t, mu_sky_t, mu_pb_t, n_sky_t, c1_t, c2_t, f_atten_t = params_test
    
    # Calculer prédictions
    buildup = 1 + c1_t*T_data + c2_t*T_data**2
    buildup = np.maximum(buildup, 0.1)
    D_direct = (A_t/r_data**2 * np.exp(-mu_air_t*r_data) 
               * buildup * np.exp(-mu_pb_t*T_data))
    D_skyshine = (B_t * r_data**(-n_sky_t) * np.exp(-mu_sky_t*r_data) 
                 * np.exp(-mu_pb_t*T_data/f_atten_t))
    D_pred = D_direct + D_skyshine
    
    ss_res = np.sum((D_data - D_pred)**2)
    ss_tot = np.sum((D_data - np.mean(D_data))**2)
    R2_ablated = 1 - ss_res/ss_tot
    
    delta_R2 = (R2_ref - R2_ablated) * 100  # En %
    
    impact = "🔴 CRITIQUE" if delta_R2 > 0.05 else "🟡 IMPORTANT" if delta_R2 > 0.01 else "🟢 FAIBLE"
    
    ablation_results.append({
        'Paramètre': param_name,
        'Valeur originale': f"{param_values[param_idx]:.4f}",
        'Fixé à': '0',
        'R² original': f"{R2_ref:.6f}",
        'R² sans param': f"{R2_ablated:.6f}",
        'ΔR² (%)': f"{delta_R2:.4f}",
        'Impact': impact
    })

df_ablation = pd.DataFrame(ablation_results)
print(df_ablation.to_string(index=False))

# Analyse 4 : Corrélation entre paramètres
print("\n\n📋 TABLEAU 4 : CORRÉLATIONS ENTRE PARAMÈTRES")
print("-" * 80)

# Récupérer la matrice de covariance depuis le fit
try:
    # Re-fit pour obtenir la covariance
    from scipy.optimize import curve_fit
    
    ln_D = np.log(D_data)
    
    def log_model(vars, A, B, mu_air, mu_sky, mu_pb, n_sky, c1, c2, f_atten):
        r, T = vars
        buildup_screen = 1 + c1*T + c2*T**2
        buildup_screen = np.maximum(buildup_screen, 0.1)
        D_direct = (A/r**2 * np.exp(-mu_air*r) 
                   * buildup_screen * np.exp(-mu_pb*T))
        D_skyshine = (B * r**(-n_sky) * np.exp(-mu_sky*r) 
                     * np.exp(-mu_pb*T/f_atten))
        D_total = D_direct + D_skyshine
        return np.log(D_total)
    
    initial_guess = param_values
    popt, pcov = curve_fit(log_model, (r_data, T_data), ln_D,
                          p0=initial_guess, maxfev=100000,
                          bounds=([0, 0, 0, 0, 0, 0.5, -1, -0.5, 1], 
                                 [10, 1, 0.1, 0.05, 1, 2, 1, 0.5, 10]))
    
    # Calculer matrice de corrélation
    std_devs = np.sqrt(np.diag(pcov))
    corr_matrix = pcov / np.outer(std_devs, std_devs)
    
    # Identifier les corrélations fortes (|r| > 0.7)
    print("\nCorrélations fortes (|r| > 0.7) :")
    strong_corr_found = False
    for i in range(len(param_names)):
        for j in range(i+1, len(param_names)):
            if abs(corr_matrix[i, j]) > 0.7:
                strong_corr_found = True
                sign = "↗" if corr_matrix[i, j] > 0 else "↘"
                print(f"   {param_names[i]} ↔ {param_names[j]} : r = {corr_matrix[i, j]:+.3f} {sign}")
    
    if not strong_corr_found:
        print("   ✅ Aucune corrélation forte détectée - paramètres bien découplés")
    
except Exception as e:
    print(f"   ⚠️ Impossible de calculer les corrélations : {e}")

print("\n" + "="*100)
```

```python
# ============================================================================
# MODÈLE #2 : SUPER HYBRIDE
# ============================================================================
print("\n" + "="*100)
print(" 🥈 MODÈLE #2 : SUPER HYBRIDE (9 paramètres)")
print("="*100)

# Récupérer les paramètres du vice-champion
A_sh2, k0_sh2, k1_sh2, mu_air_sh2, mu_pb_sh2, b1_sh2, b2_sh2, eps_sh2, lambda_sh2, R2_sh2, unc_sh2 = \
    fit_super_hybrid(r_data, T_data, D_data, plot=False)

param_names_sh = ['A', 'k₀', 'k₁', 'μ_air', 'μ_pb', 'b₁', 'b₂', 'ε', 'λ']
param_values_sh = [A_sh2, k0_sh2, k1_sh2, mu_air_sh2, mu_pb_sh2, b1_sh2, b2_sh2, eps_sh2, lambda_sh2]
param_uncertainties_sh = unc_sh2

print(f"\n📊 RÉFÉRENCE : R² = {R2_sh2:.6f}\n")

# Analyse 1 : Incertitude relative
print("📋 TABLEAU 1 : INCERTITUDES RELATIVES")
print("-" * 80)

sensitivity_data_sh = []
for i, (name, val, unc) in enumerate(zip(param_names_sh, param_values_sh, param_uncertainties_sh)):
    rel_unc = (unc / abs(val)) * 100 if abs(val) > 1e-10 else np.inf
    confidence = "✅ Bien déterminé" if rel_unc < 20 else "⚠️ Incertain" if rel_unc < 50 else "❌ Très incertain"
    
    sensitivity_data_sh.append({
        'Paramètre': name,
        'Valeur': f"{val:.4e}" if abs(val) < 0.01 or abs(val) > 100 else f"{val:.4f}",
        'Incertitude': f"{unc:.4e}" if abs(unc) < 0.01 or abs(unc) > 100 else f"{unc:.4f}",
        'Incertitude rel. (%)': f"{rel_unc:.1f}",
        'Confiance': confidence
    })

df_sensitivity_sh = pd.DataFrame(sensitivity_data_sh)
print(df_sensitivity_sh.to_string(index=False))

# Analyse 2 : Test de significativité
print("\n\n📋 TABLEAU 2 : TEST DE SIGNIFICATIVITÉ (peut-on fixer à 0 ?)")
print("-" * 80)

significance_tests_sh = []
for i, (name, val, unc) in enumerate(zip(param_names_sh, param_values_sh, param_uncertainties_sh)):
    is_significant = abs(val) > 2 * unc
    z_score = abs(val) / unc if unc > 1e-10 else np.inf
    
    if name in ['A', 'k₀', 'μ_air', 'μ_pb']:
        verdict = "🔒 ESSENTIEL (physique)"
        can_be_zero = "Non (physique)"
    elif is_significant and z_score > 3:
        verdict = "✅ SIGNIFICATIF"
        can_be_zero = "Non (>3σ)"
    elif is_significant:
        verdict = "⚠️ MARGINAL"
        can_be_zero = "Non (>2σ)"
    else:
        verdict = "❌ NON SIGNIFICATIF"
        can_be_zero = "Oui (<2σ)"
    
    significance_tests_sh.append({
        'Paramètre': name,
        'Valeur': f"{val:.4f}" if abs(val) < 100 else f"{val:.2e}",
        '|Valeur|/Incertitude': f"{z_score:.2f}",
        'Peut être 0 ?': can_be_zero,
        'Verdict': verdict
    })

df_significance_sh = pd.DataFrame(significance_tests_sh)
print(df_significance_sh.to_string(index=False))

# Analyse 3 : Test d'ablation
print("\n\n📋 TABLEAU 3 : TEST D'ABLATION - Impact sur R²")
print("-" * 80)

ablation_results_sh = []

# Tester k₁, b₁, b₂, ε (paramètres qui pourraient être nuls)
test_params = [(2, 'k₁', 0.0), (5, 'b₁', 0.0), (6, 'b₂', 0.0), (7, 'ε', 0.0)]

for param_idx, param_name, fixed_value in test_params:
    params_test = param_values_sh.copy()
    params_test[param_idx] = fixed_value
    
    A_t, k0_t, k1_t, mu_air_t, mu_pb_t, b1_t, b2_t, eps_t, lambda_t = params_test
    
    # Calculer prédictions
    k_eff = k0_t + k1_t*T_data
    k_eff = np.clip(k_eff, 0.5, 3)
    buildup = 1 + b1_t*T_data + b2_t*T_data**2
    buildup = np.maximum(buildup, 0.1)
    correction = 1 + eps_t * np.exp(-r_data/lambda_t)
    
    D_pred = A_t * r_data**(-k_eff) * buildup * np.exp(-mu_air_t*r_data - mu_pb_t*T_data) * np.abs(correction)
    
    ss_res = np.sum((D_data - D_pred)**2)
    ss_tot = np.sum((D_data - np.mean(D_data))**2)
    R2_ablated = 1 - ss_res/ss_tot
    
    delta_R2 = (R2_sh2 - R2_ablated) * 100
    
    impact = "🔴 CRITIQUE" if delta_R2 > 0.05 else "🟡 IMPORTANT" if delta_R2 > 0.01 else "🟢 FAIBLE"
    
    ablation_results_sh.append({
        'Paramètre': param_name,
        'Valeur originale': f"{param_values_sh[param_idx]:.4f}",
        'Fixé à': str(fixed_value),
        'R² original': f"{R2_sh2:.6f}",
        'R² sans param': f"{R2_ablated:.6f}",
        'ΔR² (%)': f"{delta_R2:.4f}",
        'Impact': impact
    })

df_ablation_sh = pd.DataFrame(ablation_results_sh)
print(df_ablation_sh.to_string(index=False))

# Analyse 4 : Corrélations
print("\n\n📋 TABLEAU 4 : CORRÉLATIONS ENTRE PARAMÈTRES")
print("-" * 80)

try:
    ln_D = np.log(D_data)
    
    def log_model_sh(vars, A, k0, k1, mu_air, mu_pb, b1, b2, eps, lambda_r):
        r, T = vars
        k_eff = k0 + k1*T
        k_eff = np.clip(k_eff, 0.5, 3)
        buildup = 1 + b1*T + b2*T**2
        buildup = np.maximum(buildup, 0.1)
        correction = 1 + eps * np.exp(-r/lambda_r)
        return (np.log(A) - k_eff*np.log(r) + np.log(buildup)
                - mu_air*r - mu_pb*T + np.log(np.abs(correction)))
    
    initial_guess_sh = param_values_sh
    popt_sh, pcov_sh = curve_fit(log_model_sh, (r_data, T_data), ln_D,
                                 p0=initial_guess_sh, maxfev=100000,
                                 bounds=([0, 0.5, -0.1, 0, 0, -2, -0.5, -5, 10], 
                                        [np.inf, 3, 0.1, 0.5, 1, 2, 0.5, 5, 2000]))
    
    std_devs_sh = np.sqrt(np.diag(pcov_sh))
    corr_matrix_sh = pcov_sh / np.outer(std_devs_sh, std_devs_sh)
    
    print("\nCorrélations fortes (|r| > 0.7) :")
    strong_corr_found = False
    for i in range(len(param_names_sh)):
        for j in range(i+1, len(param_names_sh)):
            if abs(corr_matrix_sh[i, j]) > 0.7:
                strong_corr_found = True
                sign = "↗" if corr_matrix_sh[i, j] > 0 else "↘"
                print(f"   {param_names_sh[i]} ↔ {param_names_sh[j]} : r = {corr_matrix_sh[i, j]:+.3f} {sign}")
    
    if not strong_corr_found:
        print("   ✅ Aucune corrélation forte détectée - paramètres bien découplés")
        
except Exception as e:
    print(f"   ⚠️ Impossible de calculer les corrélations : {e}")

print("\n" + "="*100)
```
